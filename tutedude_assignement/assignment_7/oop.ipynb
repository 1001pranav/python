### Task 1: Basic class & Object Creation

```python
class Product:
    def __init__(self, name: str, price: float, category: str):
        self.name = name
        self.price = price
        self.category = category
    
    def get_info(self):
        print(f' Name: {self.name}')
        print(f' Price: {self.price}')
        print(f' Category: {self.category}')
    
    def apply_discount(self, percent: float):
        if percent < 0 or percent > 100:
            raise ValueError("Price should not be less than 0 or more than 100 ")
        discount = (self.price) - (self.price * (percent)/100) 
        print(" Price after discount: ",discount)
        return discount
```

```python
apple = Product('Apple', 29.55, 'Groceries')
```

```python
book = Product('Book', 12, 'Stationary')
```

```python
apple.get_info()
```

```python
book.get_info()
```

```python
pen = Product('Pen', 10, 'stationary')
```

```python
final_price = pen.apply_discount(10)
```

### Task 2 Constructor & Encapsulation

```python
class product_management(Product):
    def __init__(self, name, price, category):
        super().__init__(name, price, category)
        del self.price
        self.__price = price

    def get_price(self):
        return self.__price
    
    def set_price(self, price: float):
        if price >= 0:
            raise ValueError("Value should be greater than 0")
        self.__price = price

```

### Task 3: Single Inheritance

```python
class ElectronicProduct(Product):
    def __init__(self, name, price, category, warranty_years):
        super().__init__(name, price, category) #inheritance
        self.warranty_years = warranty_years

    def get_info(self): # Overriding
        super().get_info()
        print(f' Warranty Period (in Years): {self.warranty_years}')
```

### Task 4: Polymorphism

```python
class Mobile(Product):
    def __init__(self, name, price, category):
        super().__init__(name, price, category)

    def get_info(self):
        print('You are purchasing Mobile with below')
        super().get_info()
```

```python
class Laptop(Product):
    def __init__(self, name, price, category):
        super().__init__(name, price, category)
        
    def get_info(self):
        print("You are purchasing Laptop")
        super().get_info()
```

### Task 5: Abstraction

```python
from abc import ABC, abstractmethod
```

```python
class Payments(ABC):
    @abstractmethod
    def process_payment(self): ...
```

```python
class CreditCardPayment(Payments):
    def process_payment(self):
        print("Processing Credit Card Payments")
```

```python
class UPIPayment(Payments):
    def process_payment(self):
        print("Processing UPI Payments")
```

```python
cc_payment = CreditCardPayment()
cc_payment.process_payment()
```

```python
upi_payment = UPIPayment()
upi_payment.process_payment()
```

```python
payments = Payments()
payments.process_payment()
```

### Task 6: Magic Method & operator Overloading

```python
class Product_magic(Product):
    def __init__(self, name, price, category):
        super().__init__(name, price, category)
    def __str__(self):
        return f'Product({self.name}, {self.price}, {self.category})'
    def __add__(self, other: Product):
        return self.price + other.price
    
```

```python
laptop = Product_magic('M1', 1233432423, 'Electronics')
laptop.__str__()
```

```python
mobile = Product_magic('Vivo V1', 123321, 'Electronics')
```

```python
mobile.__add__(laptop)
```
