# Python Basics Cheatsheet

This notebook covers the fundamental concepts of Python programming.

## 1. Print Statement

The `print()` function outputs text to the console.

```python
# Basic print
print("Hello, World!")

# Print multiple values
print("Hello", "World", "Python")

# Print with separator
print("Hello", "World", sep="-")

# Print without newline
print("Hello ", end="")
print("World")

# Print with formatting
name = "Python"
version = 3.12
print(f"Language: {name}, Version: {version}")  # f-string (recommended)
print("Language: {}, Version: {}".format(name, version))  # format method
print("Language: %s, Version: %.2f" % (name, version))  # % formatting (old style)
```

## 3. Variables

Variables are containers for storing data. Python is dynamically typed.

```python
# Variable assignment
x = 10
name = "Alice"
pi = 3.14159

# Multiple assignment
a, b, c = 1, 2, 3
print(a, b, c)

# Same value to multiple variables
x = y = z = 0
print(x, y, z)

# Swapping variables
a, b = 10, 20
a, b = b, a
print(f"After swap: a={a}, b={b}")

# Variable naming rules
my_variable = 1      # snake_case (recommended for variables)
myVariable = 2       # camelCase
_private = 3         # starts with underscore (convention for private)
CONSTANT = 100       # ALL_CAPS (convention for constants)

# Check variable type
print(type(x))       # <class 'int'>
print(type(name))    # <class 'str'>
```

## 4. Data Types

Python has various built-in data types.

```python
# Numeric Types
integer_val = 42                    # int
float_val = 3.14                    # float
complex_val = 3 + 4j                # complex

print(f"Integer: {integer_val}, Type: {type(integer_val)}")
print(f"Float: {float_val}, Type: {type(float_val)}")
print(f"Complex: {complex_val}, Type: {type(complex_val)}")
```

```python
# Text Type
string_val = "Hello, World!"        # str
print(f"String: {string_val}, Type: {type(string_val)}")
```

```python
# Sequence Types
list_val = [1, 2, 3, 4, 5]          # list (mutable)
tuple_val = (1, 2, 3, 4, 5)         # tuple (immutable)
range_val = range(5)                # range

print(f"List: {list_val}, Type: {type(list_val)}")
print(f"Tuple: {tuple_val}, Type: {type(tuple_val)}")
print(f"Range: {list(range_val)}, Type: {type(range_val)}")
```

```python
# Mapping Type
dict_val = {"name": "Alice", "age": 25}  # dict
print(f"Dictionary: {dict_val}, Type: {type(dict_val)}")
```

```python
# Set Types
set_val = {1, 2, 3, 4, 5}           # set (mutable, unique values)
frozenset_val = frozenset({1, 2, 3})  # frozenset (immutable)

print(f"Set: {set_val}, Type: {type(set_val)}")
print(f"Frozenset: {frozenset_val}, Type: {type(frozenset_val)}")
```

```python
# Boolean Type
bool_true = True
bool_false = False
print(f"Boolean: {bool_true}, Type: {type(bool_true)}")
```

```python
# Binary Types
bytes_val = b"Hello"                # bytes (immutable)
bytearray_val = bytearray(5)        # bytearray (mutable)
memoryview_val = memoryview(bytes(5))  # memoryview

print(f"Bytes: {bytes_val}, Type: {type(bytes_val)}")
print(f"Bytearray: {bytearray_val}, Type: {type(bytearray_val)}")
print(f"Memoryview: {memoryview_val}, Type: {type(memoryview_val)}")
```

```python
# None Type
none_val = None
print(f"None: {none_val}, Type: {type(none_val)}")
```

## 5. Type Conversion (Casting)

Convert between different data types.

```python
# String to Integer
str_num = "42"
int_num = int(str_num)
print(f"String '{str_num}' to int: {int_num}")

# String to Float
str_float = "3.14"
float_num = float(str_float)
print(f"String '{str_float}' to float: {float_num}")

# Integer to String
num = 100
str_val = str(num)
print(f"Integer {num} to string: '{str_val}'")

# Float to Integer (truncates decimal)
float_val = 3.99
int_val = int(float_val)
print(f"Float {float_val} to int: {int_val}")

# List to Tuple and vice versa
my_list = [1, 2, 3]
my_tuple = tuple(my_list)
back_to_list = list(my_tuple)
print(f"List to tuple: {my_tuple}")
print(f"Tuple to list: {back_to_list}")

# List to Set (removes duplicates)
list_with_dups = [1, 2, 2, 3, 3, 3]
unique_set = set(list_with_dups)
print(f"List to set: {unique_set}")
```

## 7. Number Systems

Python supports binary, octal, and hexadecimal number systems.

```python
# Binary (base 2) - prefix 0b
binary_num = 0b1010  # 10 in decimal
print(f"Binary 0b1010 = {binary_num}")

# Octal (base 8) - prefix 0o
octal_num = 0o17  # 15 in decimal
print(f"Octal 0o17 = {octal_num}")

# Hexadecimal (base 16) - prefix 0x
hex_num = 0xFF  # 255 in decimal
print(f"Hexadecimal 0xFF = {hex_num}")

# Convert decimal to other bases
decimal = 42
print(f"Decimal {decimal}:")
print(f"  Binary: {bin(decimal)}")
print(f"  Octal: {oct(decimal)}")
print(f"  Hexadecimal: {hex(decimal)}")
```

## 8. Random Numbers

Generate random numbers using the `random` module.

```python
import random

# Random float between 0 and 1
print(f"Random float [0, 1): {random.random()}")

# Random integer in range [a, b]
print(f"Random int [1, 10]: {random.randint(1, 10)}")

# Random integer in range [start, stop) with step
print(f"Random range(0, 100, 5): {random.randrange(0, 100, 5)}")

# Random float in range [a, b]
print(f"Random float [1.5, 6.5]: {random.uniform(1.5, 6.5)}")

# Random choice from sequence
colors = ["red", "green", "blue", "yellow"]
print(f"Random choice: {random.choice(colors)}")

# Random sample (multiple unique elements)
print(f"Random sample (2 items): {random.sample(colors, 2)}")

# Shuffle list in place
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(f"Shuffled list: {numbers}")

# Set seed for reproducibility
random.seed(42)
print(f"With seed 42: {random.random()}")
```

## 9. Math Operations

Built-in math functions and the `math` module.

```python
import math

# Built-in functions
print(f"abs(-5): {abs(-5)}")
print(f"round(3.7): {round(3.7)}")
print(f"round(3.14159, 2): {round(3.14159, 2)}")
print(f"pow(2, 3): {pow(2, 3)}")
print(f"min(1, 2, 3): {min(1, 2, 3)}")
print(f"max(1, 2, 3): {max(1, 2, 3)}")
print(f"sum([1, 2, 3, 4, 5]): {sum([1, 2, 3, 4, 5])}")

print("\n--- Math module ---")
# Math module functions
print(f"math.pi: {math.pi}")
print(f"math.e: {math.e}")
print(f"math.sqrt(16): {math.sqrt(16)}")
print(f"math.ceil(3.2): {math.ceil(3.2)}")
print(f"math.floor(3.8): {math.floor(3.8)}")
print(f"math.factorial(5): {math.factorial(5)}")
print(f"math.log(100, 10): {math.log(100, 10)}")
print(f"math.sin(math.pi/2): {math.sin(math.pi/2)}")
print(f"math.cos(0): {math.cos(0)}")
```
