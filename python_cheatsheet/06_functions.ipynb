# Python Functions Cheatsheet

Functions are reusable blocks of code that perform specific tasks.

## 1. Defining Functions

```python
# Basic function definition
def greet():
    print("Hello, World!")

greet()
```

```python
# Function with parameters
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
```

```python
# Function with return value
def add(a, b):
    return a + b

result = add(3, 5)
print(f"Sum: {result}")
```

```python
# Multiple return values
def get_stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

minimum, maximum, average = get_stats([1, 2, 3, 4, 5])
print(f"Min: {minimum}, Max: {maximum}, Avg: {average}")
```

```python
# Docstrings
def calculate_area(length, width):
    """
    Calculate the area of a rectangle.
    
    Args:
        length (float): The length of the rectangle.
        width (float): The width of the rectangle.
    
    Returns:
        float: The area of the rectangle.
    """
    return length * width

print(calculate_area(5, 3))
print(calculate_area.__doc__)
```

## 2. Function Arguments

```python
# Default arguments
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Alice")              # Uses default
greet("Bob", "Hi")          # Override default
```

```python
# Keyword arguments
def describe_person(name, age, city):
    print(f"{name} is {age} years old and lives in {city}")

# Positional arguments
describe_person("Alice", 25, "NYC")

# Keyword arguments (any order)
describe_person(city="LA", name="Bob", age=30)

# Mixed (positional first, then keyword)
describe_person("Charlie", city="Chicago", age=35)
```

```python
# *args - Variable positional arguments
def sum_all(*args):
    print(f"Args: {args}, Type: {type(args)}")
    return sum(args)

print(f"Sum: {sum_all(1, 2, 3)}")
print(f"Sum: {sum_all(1, 2, 3, 4, 5)}")
```

```python
# **kwargs - Variable keyword arguments
def print_info(**kwargs):
    print(f"Kwargs: {kwargs}, Type: {type(kwargs)}")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

print_info(name="Alice", age=25, city="NYC")
```

```python
# Combining all argument types
def func(a, b, *args, c=10, **kwargs):
    print(f"a: {a}")
    print(f"b: {b}")
    print(f"args: {args}")
    print(f"c: {c}")
    print(f"kwargs: {kwargs}")

func(1, 2, 3, 4, 5, c=20, x=100, y=200)
```

```python
# Keyword-only arguments (after *)
def func(a, b, *, c, d=10):
    """c and d must be passed as keyword arguments"""
    print(f"a={a}, b={b}, c={c}, d={d}")

func(1, 2, c=3)        # d uses default
func(1, 2, c=3, d=4)   # All specified
```

```python
# Positional-only arguments (before /)
def func(a, b, /, c, d):
    """a and b must be passed positionally"""
    print(f"a={a}, b={b}, c={c}, d={d}")

func(1, 2, 3, 4)        # All positional
func(1, 2, c=3, d=4)    # c, d as keyword
```

```python
# Combined positional-only and keyword-only
def func(a, b, /, c, *, d, e):
    """a, b: positional-only; c: either; d, e: keyword-only"""
    print(f"a={a}, b={b}, c={c}, d={d}, e={e}")

func(1, 2, 3, d=4, e=5)
func(1, 2, c=3, d=4, e=5)
```

## 3. Lambda Functions

Anonymous functions defined in a single line.

```python
# Basic lambda
square = lambda x: x ** 2
print(f"Square of 5: {square(5)}")

# Multiple arguments
add = lambda a, b: a + b
print(f"Add 3, 5: {add(3, 5)}")

# With default argument
greet = lambda name, greeting="Hello": f"{greeting}, {name}!"
print(greet("Alice"))
print(greet("Bob", "Hi"))
```

```python
# Lambda with built-in functions
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

# sorted with key
print(f"Sorted: {sorted(numbers)}")
print(f"Sorted desc: {sorted(numbers, reverse=True)}")

# Sort by custom key
words = ["banana", "pie", "apple", "cherry"]
print(f"By length: {sorted(words, key=lambda x: len(x))}")
print(f"By last char: {sorted(words, key=lambda x: x[-1])}")
```

```python
# Lambda with map, filter, reduce
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# map - apply function to each element
squared = list(map(lambda x: x ** 2, numbers))
print(f"Squared: {squared}")

# filter - keep elements where function returns True
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Evens: {evens}")

# reduce - apply function cumulatively
product = reduce(lambda a, b: a * b, numbers)
print(f"Product: {product}")
```

## 4. Scope and Namespaces

```python
# Local vs Global scope
x = 10  # Global variable

def func():
    x = 20  # Local variable (shadows global)
    print(f"Inside func: x = {x}")

func()
print(f"Outside func: x = {x}")
```

```python
# global keyword
count = 0

def increment():
    global count
    count += 1

increment()
increment()
print(f"Count: {count}")
```

```python
# nonlocal keyword (for nested functions)
def outer():
    x = 10
    
    def inner():
        nonlocal x
        x = 20
        print(f"Inner: x = {x}")
    
    inner()
    print(f"Outer: x = {x}")

outer()
```

## 5. Closures

```python
# Closure - function that remembers its enclosing scope
def make_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)

print(f"Double 5: {double(5)}")
print(f"Triple 5: {triple(5)}")
```

```python
# Counter using closure
def make_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

counter1 = make_counter()
counter2 = make_counter()

print(f"Counter1: {counter1()}, {counter1()}, {counter1()}")
print(f"Counter2: {counter2()}, {counter2()}")
```

## 6. Recursion

```python
# Factorial using recursion
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(f"5! = {factorial(5)}")
```

```python
# Fibonacci using recursion
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(f"Fibonacci sequence: {[fibonacci(i) for i in range(10)]}")
```

```python
# Fibonacci with memoization
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci_memo(n):
    if n <= 1:
        return n
    return fibonacci_memo(n - 1) + fibonacci_memo(n - 2)

print(f"Fibonacci(30): {fibonacci_memo(30)}")
```

## 7. Higher-Order Functions

```python
# Function as argument
def apply_operation(x, y, operation):
    return operation(x, y)

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

print(f"Add: {apply_operation(3, 5, add)}")
print(f"Multiply: {apply_operation(3, 5, multiply)}")
print(f"Subtract: {apply_operation(3, 5, lambda a, b: a - b)}")
```

```python
# Function returning function
def power_factory(exponent):
    def power(base):
        return base ** exponent
    return power

square = power_factory(2)
cube = power_factory(3)

print(f"Square of 5: {square(5)}")
print(f"Cube of 5: {cube(5)}")
```

## 8. Type Hints (Python 3.5+)

```python
from typing import List, Dict, Tuple, Optional, Union, Callable

# Basic type hints
def greet(name: str) -> str:
    return f"Hello, {name}!"

# Multiple parameters
def add(a: int, b: int) -> int:
    return a + b

# Complex types
def process_items(items: List[str]) -> Dict[str, int]:
    return {item: len(item) for item in items}

# Optional (can be None)
def find_item(items: List[str], target: str) -> Optional[int]:
    try:
        return items.index(target)
    except ValueError:
        return None

# Union (multiple types)
def double(x: Union[int, float]) -> Union[int, float]:
    return x * 2

# Callable
def apply_func(func: Callable[[int, int], int], a: int, b: int) -> int:
    return func(a, b)

print(greet("Alice"))
print(process_items(["apple", "banana"]))
```

## Functions Summary

| Concept | Syntax | Description |
|---------|--------|-------------|
| Basic function | `def func():` | Define a function |
| Parameters | `def func(a, b):` | Accept arguments |
| Return | `return value` | Return a value |
| Default args | `def func(a=10):` | Default parameter value |
| *args | `def func(*args):` | Variable positional args |
| **kwargs | `def func(**kwargs):` | Variable keyword args |
| Lambda | `lambda x: x * 2` | Anonymous function |
| Global | `global var` | Access global variable |
| Nonlocal | `nonlocal var` | Access enclosing scope |
| Closure | Nested function | Remember enclosing scope |
