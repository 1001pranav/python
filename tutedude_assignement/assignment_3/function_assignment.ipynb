### Task 1 Create function for applying discount and testing with (1000, 10) and (500)

```python
def apply_discount(price: int | float, discount_percent: float | int = 5) -> float:
    if discount_percent > 60:
        discount_percent = 60
    return price - (price * (discount_percent/100))
```

```python
discount_price_1 = apply_discount(1000, 10)
```

```python
discount_price_1
```

```python
discount_price_2 = apply_discount(500)
```

### Task 2 Recursive function, with negative numbers

```python
def factorial(n: int) -> int:
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
fact_number = factorial(5)
fact_number
```

```python
fact_number = factorial(0)
fact_number
```

```python
fact_number = factorial(-3)
fact_number
```

### Task 6: Combined Utility Function

```python
apply_discount = lambda price: price * 0.1
```

```python
def filter_prices(price):
    return price > 300
```

```python
def process_prices(price: list[int]):
    discounted_prices = map(apply_discount, price)
    filtered_prices = filter(filter_prices, price)
    return (list(discounted_prices), list(filtered_prices))
```

```python
def add_price(price_list: list[int], price: int) -> list[int]: 
    return price_list.append(price)
```

```python
get_avg_price = lambda price_list: sum(price_list) / len(price_list)
```

```python
get_max_price = lambda price_list: max(price_list)
```

```python
product_prices = []
```

```python
while True:
    user_input = input("""
        1 -> Add Price
        2 -> Show Average Price
        3 -> Get Max Price
        q -> Quit
""")
    if user_input == '1':
        user_input = input("enter the Price")
        if not user_input.isdigit():
            continue
        add_price(product_prices, int(user_input))
    elif user_input == '2':
        avg_price = get_avg_price(product_prices)
        print(f"Average Price is {avg_price}")
    elif user_input == '3':
        highest_prices = get_max_price(price_list=product_prices)
        print(f" Max Price in list is {highest_prices}")
    elif user_input.upper() == 'Q':
        break;
    else: 
        print("Invalid Input Press q to quit")
    
    continue
```
