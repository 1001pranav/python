# Python Comprehensions Cheatsheet

Concise ways to create lists, dictionaries, sets, and generators.

## 1. List Comprehensions

```python
# Basic syntax: [expression for item in iterable]

# Without comprehension
squares = []
for x in range(5):
    squares.append(x ** 2)
print(f"Traditional: {squares}")

# With comprehension
squares = [x ** 2 for x in range(5)]
print(f"Comprehension: {squares}")
```

```python
# With condition: [expression for item in iterable if condition]

# Even squares
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print(f"Even squares: {even_squares}")

# Filter strings
words = ["hello", "world", "python", "is", "great"]
long_words = [w for w in words if len(w) > 4]
print(f"Long words: {long_words}")
```

```python
# If-else in expression: [expr1 if condition else expr2 for item in iterable]

numbers = [1, 2, 3, 4, 5]
labels = ["even" if x % 2 == 0 else "odd" for x in numbers]
print(f"Labels: {labels}")

# Replace negative with zero
values = [-2, -1, 0, 1, 2]
positives = [x if x > 0 else 0 for x in values]
print(f"Positives: {positives}")
```

```python
# Nested loops
# [expression for item1 in iterable1 for item2 in iterable2]

# Flatten nested list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print(f"Flattened: {flat}")

# Cartesian product
colors = ['red', 'blue']
sizes = ['S', 'M', 'L']
combos = [(c, s) for c in colors for s in sizes]
print(f"Combinations: {combos}")
```

```python
# Nested comprehension (creates nested lists)

# Create matrix
matrix = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print("Matrix:")
for row in matrix:
    print(f"  {row}")

# Transpose matrix
transposed = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print("\nTransposed:")
for row in transposed:
    print(f"  {row}")
```

## 2. Dictionary Comprehensions

```python
# Basic syntax: {key: value for item in iterable}

# Squares dictionary
squares = {x: x ** 2 for x in range(1, 6)}
print(f"Squares: {squares}")

# From two lists
keys = ['a', 'b', 'c']
values = [1, 2, 3]
combined = {k: v for k, v in zip(keys, values)}
print(f"Combined: {combined}")
```

```python
# With condition
numbers = range(10)
even_squares = {x: x ** 2 for x in numbers if x % 2 == 0}
print(f"Even squares: {even_squares}")
```

```python
# Transform existing dictionary
prices = {'apple': 0.5, 'banana': 0.25, 'orange': 0.75}

# Double prices
doubled = {k: v * 2 for k, v in prices.items()}
print(f"Doubled: {doubled}")

# Filter by value
expensive = {k: v for k, v in prices.items() if v > 0.3}
print(f"Expensive: {expensive}")

# Swap keys and values
swapped = {v: k for k, v in prices.items()}
print(f"Swapped: {swapped}")
```

```python
# Word length dictionary
words = ['hello', 'world', 'python', 'programming']
lengths = {word: len(word) for word in words}
print(f"Lengths: {lengths}")

# Character count
text = "hello world"
char_count = {char: text.count(char) for char in set(text) if char != ' '}
print(f"Char count: {char_count}")
```

## 3. Set Comprehensions

```python
# Basic syntax: {expression for item in iterable}

# Unique squares
numbers = [1, 2, 2, 3, 3, 3, 4]
unique_squares = {x ** 2 for x in numbers}
print(f"Unique squares: {unique_squares}")
```

```python
# With condition
numbers = range(20)
even_set = {x for x in numbers if x % 2 == 0}
print(f"Even set: {even_set}")
```

```python
# Unique characters
text = "hello world"
unique_chars = {char.lower() for char in text if char.isalpha()}
print(f"Unique chars: {unique_chars}")
```

```python
# Practical example: unique word lengths
words = ['apple', 'banana', 'cherry', 'date', 'elderberry', 'fig']
lengths = {len(word) for word in words}
print(f"Unique lengths: {lengths}")
```

## 4. Generator Expressions

```python
# Basic syntax: (expression for item in iterable)
# Memory efficient - generates values on demand

# Generator expression
gen = (x ** 2 for x in range(5))
print(f"Generator: {gen}")
print(f"Type: {type(gen)}")

# Iterate over generator
for value in gen:
    print(value, end=' ')
print()
```

```python
# Use with functions that accept iterables
numbers = range(1, 11)

# Sum of squares
total = sum(x ** 2 for x in numbers)
print(f"Sum of squares: {total}")

# Max value
max_val = max(x ** 2 for x in numbers)
print(f"Max square: {max_val}")

# Any/All
has_even = any(x % 2 == 0 for x in numbers)
all_positive = all(x > 0 for x in numbers)
print(f"Has even: {has_even}, All positive: {all_positive}")
```

```python
# Memory comparison
import sys

# List comprehension - stores all values
list_comp = [x ** 2 for x in range(1000)]
print(f"List size: {sys.getsizeof(list_comp)} bytes")

# Generator expression - generates on demand
gen_exp = (x ** 2 for x in range(1000))
print(f"Generator size: {sys.getsizeof(gen_exp)} bytes")
```

## 5. Practical Examples

```python
# Parse CSV-like data
data = "name,age,city\nAlice,25,NYC\nBob,30,LA\nCharlie,35,Chicago"
lines = data.split('\n')
headers = lines[0].split(',')

records = [dict(zip(headers, line.split(','))) for line in lines[1:]]
print("Records:")
for record in records:
    print(f"  {record}")
```

```python
# Filter and transform
students = [
    {'name': 'Alice', 'grade': 85},
    {'name': 'Bob', 'grade': 70},
    {'name': 'Charlie', 'grade': 92},
    {'name': 'Diana', 'grade': 78}
]

# Students who passed (grade >= 75)
passed = [s['name'] for s in students if s['grade'] >= 75]
print(f"Passed: {passed}")

# Grade mapping
grades = {s['name']: 'A' if s['grade'] >= 90 else 'B' if s['grade'] >= 80 else 'C'
          for s in students}
print(f"Grades: {grades}")
```

```python
# Clean and normalize data
raw_names = ['  Alice  ', 'BOB', 'charlie', '  DIANA']
clean_names = [name.strip().title() for name in raw_names]
print(f"Clean names: {clean_names}")

# Remove duplicates while preserving order
items = ['apple', 'banana', 'apple', 'cherry', 'banana', 'date']
unique = list(dict.fromkeys(items))  # Python 3.7+ preserves dict order
print(f"Unique (ordered): {unique}")
```

```python
# Group by
from collections import defaultdict

words = ['apple', 'banana', 'cherry', 'avocado', 'blueberry', 'apricot']

# Group by first letter
by_letter = defaultdict(list)
for word in words:
    by_letter[word[0]].append(word)
print(f"By letter: {dict(by_letter)}")

# Using comprehension (Python 3.8+ walrus operator)
# groups = {k: [w for w in words if w[0] == k] for k in set(w[0] for w in words)}
```

## Comprehensions Summary

| Type | Syntax | Result |
|------|--------|--------|
| List | `[expr for x in iter]` | `list` |
| Dict | `{k: v for x in iter}` | `dict` |
| Set | `{expr for x in iter}` | `set` |
| Generator | `(expr for x in iter)` | `generator` |

### Common Patterns

| Pattern | Example |
|---------|--------|
| Basic | `[x for x in items]` |
| Transform | `[x.upper() for x in items]` |
| Filter | `[x for x in items if x > 0]` |
| If-else | `[x if x > 0 else 0 for x in items]` |
| Nested | `[x for row in matrix for x in row]` |
