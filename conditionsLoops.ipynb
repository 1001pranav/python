**Conditions**
* If Condition -

```python
if num1 > num2:
    print("num1 is greater than num2")
```

* If Else condition

```python
if num1 < num2:
    print("num1 is less than num2")
else:
    print("num1 is greater than or equal to num2")
```

* If else if and else condition

```python
if num1 < num2:
    print("num1 is less than num2")
elif num1 == num2:
    print("num1 is equal to num2")
else:
    print("num1 is greater than num2")
```

* Nested If else 

```python
if num1 > num2:
    sub = num1 - 10
    if sub > num2:
        print("sub is greater than num2")
    else:    
        print("sub is less than num2")
```

**Loops**
* While Loops

```python
i = 0
while i < 10:
    print(i)
    i += 1
```

* While loop with continue

```python
i = 10
while i > 0:
    print(i)
    i -= 1
    if i == 5:
        continue
        print("This is won't be executed")
```

* while loop with break and else 

```python
i = 0
while i < 10:
    print(i)
    i += 1
    if i == 2:
        break
else:
    print("This is won't be executed as loop is broken")
```

* While with else

```python
i = 0
while i < 5:
    print(i)
    i += 1
else:
    print("Successfully completed loop")
```

**Loops**
* For loops with continue and break and range

```python
lists = [i for i in range(10)]
for num in lists:
    if num == 5:
        break
    if num == 3: 
        continue
    print(num * 2)
```

* Loops using string

```python
string = "Hello World"
for char in string:
    print(char)
```

**Try Except Finally**

```python
try: 
    # a = 5
    # b = 1
    b = 0
    a = a/b
except ZeroDivisionError as e:
    print("Caught an zero division exception: ", e)
except NameError as e:
    print("Caught an Named exception ", e)
except Exception as e:
    print("Caught an unknown exception: ", e)
finally:
    print("finally completed")
```
