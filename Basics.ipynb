In Variables data types are dynamically assigned 

Following are the types of the data 
| **Category**       | **Data Type**  | **Description**                                            | **Example**                |
|---------------------|----------------|------------------------------------------------------------|----------------------------|
| **Numeric Types**   | `int`          | Integer numbers                                            | `10`, `-3`                 |
|                     | `float`        | Floating-point numbers (decimals)                         | `3.14`, `-2.5`             |
|                     | `complex`      | Complex numbers with real and imaginary parts             | `3 + 4j`                   |
| **Sequence Types**  | `str`          | Text (string)                                              | `"Hello"`, `'World'`       |
|                     | `list`         | Ordered, mutable collection of items                      | `[1, 2, 3]`, `['a', 'b']` |
|                     | `tuple`        | Ordered, immutable collection of items                    | `(1, 2, 3)`, `('x', 'y')` |
|                     | `range`        | Immutable sequence of numbers                             | `range(5)`                 |
| **Mapping Type**    | `dict`         | Key-value pairs                                            | `{"key": "value"}`         |
| **Set Types**       | `set`          | Unordered collection of unique items                      | `{1, 2, 3}`                |
|                     | `frozenset`    | Immutable set                                              | `frozenset([1, 2, 3])`     |
| **Boolean Type**    | `bool`         | Boolean values (`True` or `False`)                        | `True`, `False`            |
| **Binary Types**    | `bytes`        | Immutable binary data                                      | `b'hello'`                 |
|                     | `bytearray`    | Mutable binary data                                        | `bytearray(b'hello')`      |
|                     | `memoryview`   | Exposes memory of binary objects                          | `memoryview(b'hello')`     |
| **None Type**       | `NoneType`     | Represents the absence of a value                         | `None`                     |
| **Callable Types**  | `function`     | User-defined or built-in functions                        | `def func(): ...`          |
|                     | `lambda`       | Anonymous functions                                        | `lambda x: x + 1`          |
|                     | `methods`      | Methods associated with objects                           | `object.method()`          |
| **Iterator & Generator Types** | `iterator`  | Objects implementing the iterator protocol (`__iter__`, `__next__`) | `iter([1, 2, 3])`          |
|                     | `generator`    | Lazy evaluation using `yield`                             | `def gen(): yield x`       |
| **Custom Types**    | `class`        | User-defined classes                                       | `class MyClass: ...`       |
| **Special Types**   | `Any`          | Any data type (from `typing` module)                      | `Any`                      |
|                     | `Union`        | Union of multiple types (from `typing` module)            | `Union[int, str]`          |


```python
intType = 10 # Int Data type
intVal = "10" # String Data type

print(intType, type(intType))
print(intVal, type(intVal))

floatVal = 45.7 # Floating point number
print(floatVal, type(floatVal), int(floatVal))

complexType = 1j
print(complexType, type(complexType))

#Sequence Types
listDataTypes = ["text type", "Numeric Types", "Sequence Types", "Mapping Type", "Set Types", "Boolean Type", "Binary Types", "None Type"]
print(listDataTypes, type(listDataTypes))
strType = "Test"
tupleTypes = ( (strType), (intVal, floatVal, complexType));
print(tupleTypes, type(tupleTypes))

print(range(6))

# Mapping types 
dictType = {"name": "Pranav",  "age": 24 }
print(dictType, type(dictType))

# Set types
setType = {"text type", "Numeric Types", "Sequence Types", "Mapping Type", "Set Types", "Boolean Type", "Binary Types", "None Type"}
print(setType, type(setType))

frozensetType = frozenset(setType)
print(frozensetType, type(frozensetType))

# Boolean type
boolType = True
print(boolType, type(boolType))

# Binary type
binType = 0b1010
print(binType, type(binType))

# None type
noneType = None
print(noneType, type(noneType))

# Binary Types
bytesType = b"Hello"
print(bytesType, type(bytesType))

byteArrayType = bytearray(5)
print(byteArrayType, type(byteArrayType))

memoryviewType = memoryview(bytes(5))
print(memoryviewType, type(memoryviewType))
```

Generate random numbers and random string
`random.randrange(1, 10)`, generates random number between 1 to 10, If we pass only one args then its considered as stop variable, 
From 0 to the stop variable. 

```python
import random
import string
SIZE = random.randrange(1, 40)

print(SIZE)
CHARS = string.ascii_lowercase + string.ascii_lowercase + string.digits
RANDOM_VARIABLES = ''.join(random.choice(CHARS) for _ in range(SIZE))
print(RANDOM_VARIABLES)
```

**Operators**
| **Operator Type**         | **Operator**    | **Description**                                | **Example**                 | **Output**         |
|---------------------------|-----------------|------------------------------------------------|-----------------------------|--------------------|
| **Arithmetic Operators**  | `+`             | Addition                                       | `10 + 5`                    | `15`              |
|                           | `-`             | Subtraction                                    | `10 - 5`                    | `5`               |
|                           | `*`             | Multiplication                                 | `10 * 5`                    | `50`              |
|                           | `/`             | Division                                       | `10 / 3`                    | `3.333...`        |
|                           | `//`            | Floor Division                                 | `10 // 3`                   | `3`               |
|                           | `%`             | Modulus (Remainder)                            | `10 % 3`                    | `1`               |
|                           | `**`            | Exponentiation                                 | `2 ** 3`                    | `8`               |
| **Comparison Operators**  | `==`            | Equal to                                       | `10 == 5`                   | `False`           |
|                           | `!=`            | Not equal to                                   | `10 != 5`                   | `True`            |
|                           | `>`             | Greater than                                   | `10 > 5`                    | `True`            |
|                           | `<`             | Less than                                      | `10 < 5`                    | `False`           |
|                           | `>=`            | Greater than or equal to                       | `10 >= 5`                   | `True`            |
|                           | `<=`            | Less than or equal to                          | `10 <= 5`                   | `False`           |
| **Logical Operators**     | `and`           | Logical AND                                    | `True and False`            | `False`           |
|                           | `or`            | Logical OR                                     | `True or False`             | `True`            |
|                           | `not`           | Logical NOT                                    | `not True`                  | `False`           |
| **Bitwise Operators**     | `&`             | Bitwise AND                                    | `5 & 3`                     | `1`               |
|                           | `|`             | Bitwise OR                                     | `5 | 3`                      | `7`               |
|                           | `^`             | Bitwise XOR                                    | `5 ^ 3`                     | `6`               |
|                           | `~`             | Bitwise NOT                                    | `~5`                        | `-6`              |
|                           | `<<`            | Left Shift                                     | `5 << 1`                    | `10`              |
|                           | `>>`            | Right Shift                                    | `5 >> 1`                    | `2`               |
| **Assignment Operators**  | `=`             | Assign value                                   | `x = 5`                     | `x = 5`           |
|                           | `+=`            | Add and assign                                 | `x += 5` (if `x = 10`)      | `x = 15`          |
|                           | `-=`            | Subtract and assign                            | `x -= 5` (if `x = 10`)      | `x = 5`           |
|                           | `*=`            | Multiply and assign                            | `x *= 5` (if `x = 10`)      | `x = 50`          |
|                           | `/=`            | Divide and assign                              | `x /= 2` (if `x = 10`)      | `x = 5.0`         |
|                           | `//=`           | Floor divide and assign                        | `x //= 2` (if `x = 10`)     | `x = 5`           |
|                           | `**=`           | Exponentiate and assign                        | `x **= 2` (if `x = 3`)      | `x = 9`           |
|                           | `%=`            | Modulus and assign                             | `x %= 3` (if `x = 10`)      | `x = 1`           |
| **Membership Operators**  | `in`            | Checks if a value is in a sequence             | `'a' in 'apple'`            | `True`            |
|                           | `not in`        | Checks if a value is not in a sequence         | `'b' not in 'apple'`        | `True`            |
| **Identity Operators**    | `is`            | Checks if two objects are identical            | `x is y`                    | `True/False`      |
|                           | `is not`        | Checks if two objects are not identical        | `x is not y`                | `True/False`      |
| **Special Operators**     | `[]`            | Indexing                                       | `lst[1]` (if `lst = [1, 2]`) | `2`              |
|                           | `:`             | Slicing                                        | `lst[0:2]`                  | `[1, 2]`          |
|                           | `,`             | Separator                                      | `a, b = 1, 2`               | `a = 1, b = 2`    |
|                           | `()`            | Function call or grouping                      | `print(5 + 3)`              | `8`               |
|                           | `{}`            | Dictionary or set                              | `{'a': 1}`                  | `{'a': 1}`        |

```python
num1, num2 = 220, 250
add = num1 + num2
sub = num1 - num2
mul = num1 * num2
div = num1 / num2
flrDiv = num1 // num2
mod = num1 % num2
expo = num1 ** num2

print("num1 = ",num1, " num2 = ",num2,
      "\nAddition = ", add, 
      "\nSubtraction = ", sub, 
      "\nMultiplication = ", mul, 
      "\nDivision = ", div, 
      "\nFloor Division = ", flrDiv, 
      "\nModulus = ", mod, 
      "\nExponentiation = ", expo)
```
