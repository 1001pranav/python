# Task 4: File Handling (Read, Write, Append, Modes)
### Writing the sales record

```python
sales = [1200, 450, 980, 1500, 3000]
```

```python
import os
```

```python
file_path = os.path.join('assets', 'sales_data.txt')
```

```python
with open(file_path, 'w') as file:
    for price in sales:
        file.write(str(price))
        file.write('\n')
    file.close()

with open(file_path, 'r') as file:
    data = file.read()
    print(data)
    file.close()
```

### Task 2 reading the files

```python
with open(file_path, 'r') as file:
    data = file.read()
    print(data)
    file.close()

with open(file_path, 'r') as file:
    data = file.readline()
    print(data)
    file.close()

convert_integer = lambda data: int(data)

with open(file_path, 'r') as file:
    data = file.readlines()
    list_data = list(map(convert_integer, data))
    print(list_data)
    file.close()
```

### Task 3: Append new Sales Data

```python
new_sales = [5000, 2500, 1700]
```

```python
with open(file_path, 'a') as file:
    for price in new_sales:
        file.write(f'{price} \n')
    file.close()
```

### Task 4: Generate Sales Summary Report

```python
convert_to_int = lambda data: int(data)
```

```python
sales_lists = []
```

```python
with open(file_path, 'r') as file:
    data = file.readlines()
    file.close()
    sales_lists = list(map(convert_to_int, data))
```

```python
sum_sales = sum(sales_lists)
```

```python
max_sales = max(sales_lists)
```

```python
min_sales = min(sales_lists)
```

```python
avg_sales = sum_sales / len(sales_lists)
```

```python
print("Total Sales: ", sum_sales)
print("Highest Sale: ", max_sales)
print("Lowest Sale: ", min_sales)
print("Average Sale", avg_sales)
```

### Task 5: Creating Product info file

```python
product_file_name = os.path.join('assets', 'products.txt')
```

```python
user_input = ['productName | Price \n' ]
```

```python
for _ in range(3):
    user_input_product_name = input("Enter Product Name")
    user_input_price_name = input("Enter Price of the product")
    user_input.append(f'{user_input_product_name} ')
    user_input.append( f'{user_input_price_name}')
    user_input.append('\n')
```

```python
with open(product_file_name, 'w') as file:
    file.writelines(user_input)
```

### Task 6: Read file Safety

```python
user_input = input("Enter The file Name: ")
```

```python
user_file_path = os.path.join("assets", user_input)
```

```python
if not os.path.exists(user_file_path):
    print("File Not Found please check file Name")
else: 
    try:
        with open(user_file_path) as file:
            file_data = file.read()
            print(file_data)
    except FileNotFoundError as E:
        print(E)
    except PermissionError as E:
        print(E)
```

```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}
```

```python
discount_price = input("Enter discount Percentage")
discount_price = int(discount_price)
```

```python
discount_price
```

```python
sales_report = ['Product | Original Price | Discounted Price \n']
total_items=0, avg_discount=0
```

```python
for key in prices:
    sales_report.append(f'{key} ')
    sales_report.append(f'{prices[key]} ')
    sales_report.append(f'{prices[key] * (discount_price/100)} \n')
```

```python
sales_report
```

```python
total_items = len(prices.values())
```

```python
avg_discount=sum(prices.values())/total_items
```

```python
sales_report.append(f"Summary: \n ----- \n Total items = {total_items} \n Average Discount = {avg_discount} \n ")
```

```python
file_path = os.path.join('assets', 'discount_report.txt')
```

```python
with open(file_path, 'w') as file:
    file.writelines(sales_report)
```
