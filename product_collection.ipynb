# Task 3 Pricing Dict
### 3.1 Product as Key and Price as Value

```python
price_dict = {}
```

```python

for item, index in zip(products, range(len(products))):
    price_dict[item] = (index + 1) * 1000
print(price_dict)
```

### Task 3.2 Create, Update, Delete Dict

#### Updating a existing price

```python
price_dict.update({"Laptop": 150000})
price_dict
```

### Create a new existing price

```python
price_dict['Calculator'] = 150
```

### If Item existing dict then delete it else ignore

```python
if "Toy" in price_dict: 
    del price_dict["Toy"] # we can also use del keyword to remove an item
if "Noodles" in price_dict:
    price_dict.pop("Noodles") # Other way to remove an item is use .pop method
price_dict
```

### Print average price of all the products 

```python
avg_price = sum(price_dict.values()) / len(price_dict)
avg_price
```

### Task 4 Combining everything into One

```python
catalogue = []
```

```python
def get_category_of_product(item: str) -> str | None:
    
    if item in ["Laptop", "Smartphone", "Calculator"]:
        category = "Electronics"
    elif item in ["Noodles", "Rice"]:
        category = "Groceries"
    elif item in ["Notebook", "Pen"]:
        category = "Stationery"
    elif item in ["Makeup kit"]:
        category = "Accessories"
    elif item in ["water Bottle"]:
        category = "Clothing"
    else:
        category = None
    return category
```

```python
for item in products:
    categories = get_category_of_product(item)
    if item in price_dict:
        catalogue.append((item, price_dict[item], categories))
    else:
        catalogue.append((item, None, categories))
print(catalogue)
```

### Task 4.2 

```python
category_to_product = {}
max_value = -1
```

```python
for item, price, category in catalogue:
    if category not in category_to_product:
        category_to_product[category] = []
    category_to_product[category].append((item, price))
print(category_to_product)
```

```python
for category, items in category_to_product.items():
    if len(items) > max_value:
        max_value = len(items)
        max_category = category
print("Category with most products:", max_category, "with", max_value, "products")
```
