`A pandas Series is very similar to a one-dimensional NumPy array, but it has additional functionality that allows values in the Series to be indexed using labels. A NumPy array does not have the flexibility to do this. This labeling is useful when you are storing pieces of data that have other data associated with them. Say you want to store the ages of students in an online course to eventually figure out the average student age. If stored in a NumPy array, you could only access these ages with the internal ndarray indices 0,1,2.... With a Series object, the indices of values are set to 0,1,2... by default, but you can customize the indices to be other values such as student names so an age can be accessed using a name.`

```python
ages = np.array([13,25,19])
series1 = pd.Series(ages)
print(series1)
```

```python
ages = np.array([13,25,19])
series1 = pd.Series(ages,index=['Emma', 'Swetha', 'Serajh'])
print(series1)
```

### Understanding Pandas DataFrame

A **DataFrame** in pandas is like a matrix with rows and columns. Here's a simple explanation:

- **Rows and Columns**: 
  - Rows and columns can be labeled with **integers** or **strings**.
- **Data Types**:
  - A DataFrame can hold different types of data (e.g., numbers, strings) in different columns.
  - However, within a column, all values must have the **same data type**.
  - Each column is essentially a **Series**.
- **Column Length**:
  - All columns must have the **same number of rows**.

### Creating a DataFrame

You can create a DataFrame in various ways:
- From a **CSV file**
- From a **SQL query**
- Using a **Python list** or a **dictionary**

```python

data = [
    [1, 'Alice', 25],
    [2, 'Bob', 30],
    [3, 'Charlie', 35]
]

# Define column names
columns = ['ID', 'Name', 'Age']

# Create the DataFrame
df = pd.DataFrame(data, columns=columns)

print(df)
```

# NumPy Built-in Functions for Statistical Analysis

NumPy provides various built-in functions for statistical analysis, ranging from basic statistical measures to advanced operations like correlation and covariance.

## **1. Basic Statistical Measures**

| **Function**   | **Description**                                 |
|-----------------|-------------------------------------------------|
| `np.mean()`     | Calculates the mean (average) of the array.    |
| `np.median()`   | Calculates the median value of the array.      |
| `np.std()`      | Calculates the standard deviation.             |
| `np.var()`      | Calculates the variance.                       |
| `np.sum()`      | Computes the sum of all elements in the array. |
| `np.prod()`     | Computes the product of all elements in the array. |
| `np.min()`      | Finds the minimum value.                       |
| `np.max()`      | Finds the maximum value.                       |
| `np.ptp()`      | Calculates the range (max - min) of values.    |


Mean

```python
arr = np.arange(1, 6)
```

```python
mean = np.mean(arr)
median = np.median(arr)
std = np.std(arr)
var = np.var(arr)

print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std)
print("Variance:", var)
```

```python
sum_arr = np.sum(arr)
prod_arr = np.prod(arr)
print("Sum of element", sum_arr)
print("Product of element", prod_arr)
```

```python
min_num = np.min(arr)
max_num = np.max(arr)
ptp = np.ptp(arr)

print("Minimum number:", min_num)
print("Maximum number:", max_num)
print("Peak-to-peak value:", ptp)
```

# Percentile and Quantile

Percentile Computes the q-th percentile of the data.  
A percentile is a value that indicates the percentage of data points in a dataset that are below it. Percentiles divide the dataset into 100 equal parts.

```python
arr = np.random.randint(1, 100, size=(11,))
```

```python
arr.sort()
arr
```

```python
# 25th percentile
p25 = np.percentile(arr, 25)
print("25th percentile:", p25)  
# 50th percentile  
p50 = np.percentile(arr, 50)
print("50th percentile:", p50)  
# 90th percentile
p90 = np.percentile(arr, 90)
print("90th percentile:", p90)
```

Quantile is similar to percentile but it uses fractions instead of percentage

```python
q22 = np.quantile(arr, 0.22)
print("22nd percentile:", q22)

q37 = np.quantile(arr, 0.37)
print("37th percentile:", q37)

q79 = np.quantile(arr, 0.79)
print("79th percentile:", q79)
```

# Broadcast and Vectorized

**Vectorized operations** are operations that are applied to entire arrays (or slices of arrays) at once, rather than using explicit loops. This is faster and more efficient because NumPy is optimized for such operations.

```python
array1 = np.array([1, 2, 3])  # Shape (3)
array2 = np.array([10, 20, 30])  # Shape (3,)
```

```python
array1 + array2 # Vectorized operation
```

# Broadcasting

Broadcasting is a powerful feature in NumPy that allows you to perform operations on arrays of different shapes. NumPy automatically "broadcasts" the smaller array across the larger one so they have compatible shapes.

Rules of Broadcasting:

1. Align shapes from the right: Compare the shapes of the two arrays from the rightmost dimension.
2. Dimensions must match or be 1: For each dimension, the sizes must either be equal, or one of them must be 1.
3. Missing dimensions are treated as 1: If one array has fewer dimensions, it’s padded with ones on the left.

```python
# Example 1: Broadcasting a scalar to an array
array1 = np.array([1, 2, 3, 4, 5]);

array1 + 5
```

```python
# Example 2: Broadcasting arrays of different shapes
array2 = np.array([[11, 22, 33, 44, 55], [10, 20, 30, 40, 50]]);

array1 + array2
```

# Vectorized Operations
Vectorized operations allow you to perform element-wise operations on entire arrays (or slices) without writing explicit loops. Instead of iterating over each element manually, NumPy applies the operation to all elements at once, leveraging optimized, low-level code (written in C) under the hood. This makes them much faster than Python loops.

1. Operations are applied to every element.
2. We don't required multiple loops for applying operations for every element.
3. Works with Scalars and Arrays

Vectorized operations are performed on similar shape arrays.
If Shapes are different, With help of the Broadcasting Vectorized operations are performed.

```python
array1 = np.random.randint(1, 10, size=(2, 5))
array2 = np.random.randint(1, 10, size=(2, 5))
array1, array2
```

```python
array1 + array2 # Vectorized operation are applied here
```

```python
np.sin(array1)
```

```python
array2 < array1 % 2
#  array1 % 2 is converting to Int instead of float
```

```python
# Vectorized operation enabled by broadcasting
matrix = np.array([[1, 2], [3, 4]])  # Shape: (2, 2)
vector = np.array([10, 20])          # Shape: (2,)

matrix + vector  
# Broadcasting stretches `vector` to [[10, 20], [10, 20]]
# Result: [[11, 22], [13, 24]]
```
