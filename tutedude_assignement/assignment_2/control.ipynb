# Task 1 Discount Rules
### take User input and add discount and Calculate taxes

```python
def get_discount(order_amount: int) -> float:
    if order_amount >= 2000:
        return 0.15
    elif order_amount >= 1000:
        return 0.10
    elif order_amount >= 500:
        return 0.05
    else:
        return 0.0
```

```python
order_input = input("Enter the order amount: ")
try:
    if not order_input.isdigit():
        raise ValueError("Input is not a valid integer.")
    order_amount = int(order_input)
    if order_amount < 0 or order_amount is None:
        raise ValueError("Invalid order amount.")
    
    discount = get_discount(order_amount)

    print (f"Sub Total: {order_amount}");
    print (f"discount applied: {discount * 100}%");
    print (f"Sub Total after discount: {order_amount * (1 - discount)}");
    print (f"Tax: {100 * 0.05}");
    print (f"Total: {(order_amount * (1 - discount)) + (100 * 0.05)}");
    
except ValueError as e:
    print(e)
    exit(1)
```

# Task 2 Process multiple orders
### Using For loop for order processing

```python
orders = [1200, 2500, 800, 1750, 3000]
total_amount = 0
total_discount = 0
```

```python
for order_amount in orders:
    discount = get_discount(order_amount)
    if discount > 0.0:
        total_discount += 1
    
    total = (order_amount * (1 - discount))
    print(f" Order Amount: {order_amount} ")
    print(f" Discount applied: {discount * 100}% ")
    print(f" Sub Total: {total} \n")
    print(f" Total after tax: {total + (total * 0.05)} \n")
    total_amount += total
print(f"\nTotal number of orders with discount >0 applied: {total_discount}")
print(f"Total amount for all orders after discounts: {total_amount}")
```

# Task 3 Use of Continue/Break
Demonstration of Continue and Break 

```python
user_order_amount = []
```

```python
while True:
    user_input = input("""
    1 - Add order amount
    2 - Show all orders after total calculation
    q - Quit
    Enter your choice: """)
    if user_input == '1':
        order_input = input("Enter the order amount: ")
        try:
            if not order_input.isdigit():
                raise ValueError("Input is not a valid integer.")
            order_amount = int(order_input)
            if order_amount < 0 or order_amount is None:
                raise ValueError("Invalid order amount.")
            user_order_amount.append(order_amount)
            continue
        except ValueError as e:
            print(e)
            continue
    elif user_input == '2':
        total_amount = 0
        for order_amount in user_order_amount:
            total_amount += order_amount
            print(f" Order Amount: {order_amount} ")
            continue
    elif user_input == 'q' or user_input == 'Q':
        break
    else:
        print("Invalid choice. Please try again.")
        continue
```

# Task 5 For Loops with break and continue
Using Break and continue with for Loops

```python
daily = [200, 150, 0, 400, 50, -1, 300]
total_sales = 0
```

```python
for order in daily:
    if order == 0:
        continue
    elif order == -1:
        break
    else:
        total_sales += order
```

```python
print(f"Total sales {total_sales}")
```
