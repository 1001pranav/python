# Python Exception Handling Cheatsheet

Handling errors gracefully in Python programs.

## 1. Basic Try-Except

````python
# Basic try-except
try:
    result = 10 / 0
except:
    print("An error occurred!")
````

````python
# Catch specific exception
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
````

````python
# Access exception details
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"Error type: {type(e).__name__}")
    print(f"Error message: {e}")
````

## 2. Multiple Exceptions

````python
# Multiple except blocks
def process_input(value):
    try:
        number = int(value)
        result = 100 / number
        return result
    except ValueError:
        print("Invalid input: not a number")
    except ZeroDivisionError:
        print("Invalid input: cannot be zero")

process_input("abc")  # ValueError
process_input("0")    # ZeroDivisionError
print(process_input("5"))  # Success
````

````python
# Catch multiple exceptions in one block
try:
    # Could raise ValueError or ZeroDivisionError
    result = int("abc")
except (ValueError, ZeroDivisionError) as e:
    print(f"Caught: {type(e).__name__}: {e}")
````

````python
# Catch all exceptions (use with caution)
try:
    result = 10 / 0
except Exception as e:
    print(f"Caught exception: {type(e).__name__}: {e}")
````

## 3. Try-Except-Else-Finally

````python
# else - runs if no exception
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero!")
    else:
        print(f"Result: {result}")
        return result

divide(10, 2)
divide(10, 0)
````

````python
# finally - always runs
def process_file(filename):
    file = None
    try:
        file = open(filename, 'r')
        content = file.read()
        return content
    except FileNotFoundError:
        print(f"File '{filename}' not found")
    finally:
        if file:
            file.close()
            print("File closed")
        print("Finally block executed")

process_file("nonexistent.txt")
````

````python
# Complete try-except-else-finally
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Division by zero")
        return None
    except TypeError:
        print("Error: Invalid types")
        return None
    else:
        print(f"Success: {a} / {b} = {result}")
        return result
    finally:
        print("Operation complete\n")

safe_divide(10, 2)
safe_divide(10, 0)
safe_divide("10", 2)
````

## 4. Raising Exceptions

````python
# Raise an exception
def validate_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    if age > 150:
        raise ValueError("Age seems unrealistic")
    return True

try:
    validate_age(-5)
except ValueError as e:
    print(f"Validation error: {e}")
````

````python
# Re-raise exception
def process_data(data):
    try:
        # Some processing
        if not data:
            raise ValueError("Data is empty")
    except ValueError:
        print("Logging error...")
        raise  # Re-raise the exception

try:
    process_data(None)
except ValueError as e:
    print(f"Caught re-raised exception: {e}")
````

````python
# Raise from (exception chaining)
def fetch_data():
    try:
        # Simulate an error
        raise ConnectionError("Failed to connect to server")
    except ConnectionError as e:
        raise RuntimeError("Data fetch failed") from e

try:
    fetch_data()
except RuntimeError as e:
    print(f"Error: {e}")
    print(f"Caused by: {e.__cause__}")
````

## 5. Custom Exceptions

````python
# Simple custom exception
class CustomError(Exception):
    pass

try:
    raise CustomError("Something went wrong")
except CustomError as e:
    print(f"Caught CustomError: {e}")
````

````python
# Custom exception with attributes
class ValidationError(Exception):
    def __init__(self, field, message):
        self.field = field
        self.message = message
        super().__init__(f"{field}: {message}")

def validate_user(name, age):
    if not name:
        raise ValidationError("name", "Name is required")
    if age < 0:
        raise ValidationError("age", "Age must be positive")

try:
    validate_user("", 25)
except ValidationError as e:
    print(f"Field: {e.field}")
    print(f"Message: {e.message}")
````

````python
# Exception hierarchy
class ApplicationError(Exception):
    """Base exception for application"""
    pass

class DatabaseError(ApplicationError):
    """Database-related errors"""
    pass

class NetworkError(ApplicationError):
    """Network-related errors"""
    pass

try:
    raise DatabaseError("Connection failed")
except ApplicationError as e:
    print(f"Application error: {type(e).__name__}: {e}")
````

## 6. Common Built-in Exceptions

````python
# ValueError - invalid value
try:
    int("abc")
except ValueError as e:
    print(f"ValueError: {e}")
````

````python
# TypeError - wrong type
try:
    "hello" + 5
except TypeError as e:
    print(f"TypeError: {e}")
````

````python
# KeyError - key not in dict
try:
    d = {"a": 1}
    value = d["b"]
except KeyError as e:
    print(f"KeyError: {e}")
````

````python
# IndexError - index out of range
try:
    lst = [1, 2, 3]
    value = lst[10]
except IndexError as e:
    print(f"IndexError: {e}")
````

````python
# AttributeError - attribute not found
try:
    "hello".nonexistent()
except AttributeError as e:
    print(f"AttributeError: {e}")
````

````python
# NameError - name not defined
try:
    print(undefined_variable)
except NameError as e:
    print(f"NameError: {e}")
````

````python
# FileNotFoundError
try:
    with open("nonexistent_file.txt") as f:
        pass
except FileNotFoundError as e:
    print(f"FileNotFoundError: {e}")
````

## 7. Assert Statements

````python
# Basic assert
def calculate_average(numbers):
    assert len(numbers) > 0, "List cannot be empty"
    return sum(numbers) / len(numbers)

print(calculate_average([1, 2, 3, 4, 5]))

try:
    calculate_average([])
except AssertionError as e:
    print(f"AssertionError: {e}")
````

````python
# Assert for type checking (debugging)
def greet(name):
    assert isinstance(name, str), f"Expected str, got {type(name).__name__}"
    return f"Hello, {name}!"

print(greet("Alice"))

try:
    greet(123)
except AssertionError as e:
    print(f"AssertionError: {e}")
````

## 8. Context Managers for Error Handling

````python
# suppress specific exceptions
from contextlib import suppress

# Without suppress
# try:
#     os.remove('file.txt')
# except FileNotFoundError:
#     pass

# With suppress
import os
with suppress(FileNotFoundError):
    os.remove('nonexistent_file.txt')
    
print("No error raised!")
````

````python
# Custom context manager for error handling
from contextlib import contextmanager

@contextmanager
def error_handler(error_message):
    try:
        yield
    except Exception as e:
        print(f"{error_message}: {e}")

with error_handler("Math error"):
    result = 10 / 0
````

## 9. Best Practices

````python
# 1. Be specific with exceptions
# Bad:
# try:
#     do_something()
# except:
#     pass

# Good:
# try:
#     do_something()
# except ValueError as e:
#     handle_value_error(e)
# except TypeError as e:
#     handle_type_error(e)

# 2. Don't use exceptions for flow control
# Bad:
# try:
#     value = my_list[index]
# except IndexError:
#     value = default

# Good:
# if index < len(my_list):
#     value = my_list[index]
# else:
#     value = default

# 3. Use EAFP (Easier to Ask Forgiveness than Permission)
# LBYL (Look Before You Leap) - less Pythonic:
# if key in dictionary:
#     value = dictionary[key]

# EAFP - more Pythonic:
# try:
#     value = dictionary[key]
# except KeyError:
#     value = default

# Or use dict.get():
# value = dictionary.get(key, default)

print("Best practices examples (see comments)")
````
