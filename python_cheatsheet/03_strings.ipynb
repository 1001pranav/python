# Python Strings Cheatsheet

Strings are sequences of characters used to store and manipulate text.

## 1. Creating Strings

```python
# Single quotes
str1 = 'Hello, World!'

# Double quotes
str2 = "Hello, World!"

# Triple quotes (multi-line)
str3 = '''This is a
multi-line
string'''

str4 = """Another way to create
multi-line strings"""

# Raw strings (ignores escape characters)
raw_str = r"C:\Users\name\folder"

print(str1)
print(str3)
print(f"Raw string: {raw_str}")
```

## 2. String Indexing and Slicing

```python
text = "Python Programming"

# Indexing (0-based)
print(f"First character: {text[0]}")       # P
print(f"Last character: {text[-1]}")       # g
print(f"5th character: {text[4]}")         # o

# Slicing [start:end:step]
print(f"\nFirst 6 chars: {text[:6]}")      # Python
print(f"From index 7: {text[7:]}")         # Programming
print(f"Index 7-14: {text[7:14]}")         # Program
print(f"Every 2nd char: {text[::2]}")      # Pto rgamn
print(f"Reverse: {text[::-1]}")            # gnimmargorP nohtyP
print(f"Last 5 chars: {text[-5:]}")        # mming
```

## 3. String Methods - Case Conversion

```python
text = "hello WORLD python"

print(f"Original: {text}")
print(f"upper(): {text.upper()}")           # HELLO WORLD PYTHON
print(f"lower(): {text.lower()}")           # hello world python
print(f"capitalize(): {text.capitalize()}") # Hello world python
print(f"title(): {text.title()}")           # Hello World Python
print(f"swapcase(): {text.swapcase()}")     # HELLO world PYTHON
print(f"casefold(): {text.casefold()}")     # hello world python (aggressive lowercase)
```

## 4. String Methods - Search and Find

```python
text = "Hello, World! Welcome to Python World!"

# Find substring (returns index, -1 if not found)
print(f"find('World'): {text.find('World')}")         # 7
print(f"find('Python'): {text.find('Python')}")       # 25
print(f"find('Java'): {text.find('Java')}")           # -1

# rfind - finds last occurrence
print(f"rfind('World'): {text.rfind('World')}")       # 32

# index - like find but raises ValueError if not found
print(f"index('World'): {text.index('World')}")       # 7

# count occurrences
print(f"count('World'): {text.count('World')}")       # 2
print(f"count('o'): {text.count('o')}")               # 5

# Check start/end
print(f"startswith('Hello'): {text.startswith('Hello')}")  # True
print(f"endswith('!'): {text.endswith('!')}")              # True
```

## 5. String Methods - Modify

```python
text = "   Hello, World!   "

# Strip whitespace
print(f"Original: '{text}'")
print(f"strip(): '{text.strip()}'")
print(f"lstrip(): '{text.lstrip()}'")
print(f"rstrip(): '{text.rstrip()}'")

# Replace
print(f"\nreplace('World', 'Python'): {text.strip().replace('World', 'Python')}")

# Remove specific characters
text2 = "###Hello###"
print(f"strip('#'): {text2.strip('#')}")

# Center, ljust, rjust
word = "Python"
print(f"\ncenter(20, '-'): '{word.center(20, '-')}'")
print(f"ljust(20, '-'): '{word.ljust(20, '-')}'")
print(f"rjust(20, '-'): '{word.rjust(20, '-')}'")
print(f"zfill(10): '{word.zfill(10)}'")
```

## 6. String Methods - Split and Join

```python
# Split
text = "apple,banana,cherry,date"
fruits = text.split(",")
print(f"split(','): {fruits}")  # ['apple', 'banana', 'cherry', 'date']

# Split with maxsplit
print(f"split(',', 2): {text.split(',', 2)}")

# Split by whitespace
sentence = "Hello   World   Python"
words = sentence.split()  # splits by any whitespace
print(f"split(): {words}")

# splitlines
multiline = "Line1\nLine2\nLine3"
print(f"splitlines(): {multiline.splitlines()}")

# Join
words = ['Hello', 'World', 'Python']
print(f"' '.join(): {' '.join(words)}")
print(f"'-'.join(): {'-'.join(words)}")
print(f"''.join(): {''.join(words)}")

# Partition
text = "Hello World Python"
print(f"partition(' '): {text.partition(' ')}")
print(f"rpartition(' '): {text.rpartition(' ')}")
```

## 7. String Methods - Check Type

```python
# isalpha - only letters
print(f"'Hello'.isalpha(): {'Hello'.isalpha()}")
print(f"'Hello123'.isalpha(): {'Hello123'.isalpha()}")

# isdigit - only digits
print(f"\n'12345'.isdigit(): {'12345'.isdigit()}")
print(f"'123.45'.isdigit(): {'123.45'.isdigit()}")

# isalnum - letters and digits
print(f"\n'Hello123'.isalnum(): {'Hello123'.isalnum()}")
print(f"'Hello 123'.isalnum(): {'Hello 123'.isalnum()}")

# isspace - only whitespace
print(f"\n'   '.isspace(): {'   '.isspace()}")

# isupper, islower
print(f"\n'HELLO'.isupper(): {'HELLO'.isupper()}")
print(f"'hello'.islower(): {'hello'.islower()}")

# istitle
print(f"\n'Hello World'.istitle(): {'Hello World'.istitle()}")

# isnumeric, isdecimal
print(f"\n'123'.isnumeric(): {'123'.isnumeric()}")
print(f"'\u00B2'.isnumeric(): {'\u00B2'.isnumeric()}")  # superscript 2

# isidentifier - valid variable name
print(f"\n'myVar'.isidentifier(): {'myVar'.isidentifier()}")
print(f"'2var'.isidentifier(): {'2var'.isidentifier()}")
```

## 8. String Formatting

```python
name = "Alice"
age = 25
price = 49.99

# f-strings (Python 3.6+) - Recommended
print(f"Name: {name}, Age: {age}")
print(f"Price: ${price:.2f}")
print(f"Debug: {name=}, {age=}")  # Python 3.8+

# format() method
print("Name: {}, Age: {}".format(name, age))
print("Name: {0}, Age: {1}, Again: {0}".format(name, age))
print("Name: {n}, Age: {a}".format(n=name, a=age))

# % formatting (old style)
print("Name: %s, Age: %d" % (name, age))
print("Price: %.2f" % price)

# Number formatting
num = 1234567.89
print(f"\nNumber formatting:")
print(f"Default: {num}")
print(f"With comma: {num:,}")
print(f"2 decimals: {num:.2f}")
print(f"Scientific: {num:e}")
print(f"Percentage: {0.85:.1%}")

# Alignment
text = "Hi"
print(f"\nAlignment:")
print(f"Left:   '{text:<10}'")
print(f"Right:  '{text:>10}'")
print(f"Center: '{text:^10}'")
print(f"Fill:   '{text:*^10}'")
```

## 9. Escape Characters

```python
# Common escape characters
print("Newline:\nSecond line")
print("Tab:\tIndented")
print("Backslash: \\")
print("Single quote: \'")
print("Double quote: \"")
print("Carriage return: Hello\rWorld")  # World replaces Hello
print("Backspace: Hello\b World")        # Hell World

# Unicode
print(f"\nUnicode: \u0048\u0065\u006C\u006C\u006F")  # Hello
print(f"Emoji: \U0001F600")  # Grinning face
```

## 10. String Operations

```python
# Concatenation
s1 = "Hello"
s2 = "World"
print(f"Concatenation: {s1 + ' ' + s2}")

# Repetition
print(f"Repetition: {s1 * 3}")

# Length
print(f"Length: {len(s1)}")

# Membership
print(f"'ell' in 'Hello': {'ell' in 'Hello'}")

# Iteration
print("Characters:", end=" ")
for char in "Python":
    print(char, end=" ")
print()

# min, max, sorted
text = "python"
print(f"\nmin('{text}'): {min(text)}")
print(f"max('{text}'): {max(text)}")
print(f"sorted('{text}'): {sorted(text)}")

# ord and chr
print(f"\nord('A'): {ord('A')}")
print(f"chr(65): {chr(65)}")
```

## 11. String Translation

```python
# maketrans and translate
text = "Hello World"

# Create translation table
table = str.maketrans("aeiou", "12345")  # vowels to numbers
translated = text.translate(table)
print(f"Original: {text}")
print(f"Translated: {translated}")

# Remove characters
table = str.maketrans("", "", "aeiou")  # remove vowels
print(f"Remove vowels: {text.translate(table)}")

# encode and decode
text = "Hello, World!"
encoded = text.encode('utf-8')
decoded = encoded.decode('utf-8')
print(f"\nEncoded: {encoded}")
print(f"Decoded: {decoded}")
```

## String Methods Summary Table

| Method | Description | Example |
|--------|-------------|----------|
| `upper()` | Uppercase | `"hello".upper()` → `"HELLO"` |
| `lower()` | Lowercase | `"HELLO".lower()` → `"hello"` |
| `capitalize()` | First char upper | `"hello".capitalize()` → `"Hello"` |
| `title()` | Title case | `"hello world".title()` → `"Hello World"` |
| `strip()` | Remove whitespace | `" hi ".strip()` → `"hi"` |
| `split()` | Split to list | `"a,b,c".split(",")` → `["a","b","c"]` |
| `join()` | Join list | `",".join(["a","b"])` → `"a,b"` |
| `replace()` | Replace substring | `"hi".replace("h","H")` → `"Hi"` |
| `find()` | Find index | `"hello".find("l")` → `2` |
| `count()` | Count occurrences | `"hello".count("l")` → `2` |
| `startswith()` | Check start | `"hello".startswith("he")` → `True` |
| `endswith()` | Check end | `"hello".endswith("lo")` → `True` |
| `isalpha()` | Only letters | `"hello".isalpha()` → `True` |
| `isdigit()` | Only digits | `"123".isdigit()` → `True` |
| `isalnum()` | Letters/digits | `"hi123".isalnum()` → `True` |
