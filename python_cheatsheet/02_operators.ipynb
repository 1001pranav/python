# Python Operators Cheatsheet

Operators are used to perform operations on variables and values.

## 1. Arithmetic Operators

Used for mathematical calculations.

```python
a, b = 15, 4

print(f"a = {a}, b = {b}")
print(f"Addition (a + b): {a + b}")           # 19
print(f"Subtraction (a - b): {a - b}")        # 11
print(f"Multiplication (a * b): {a * b}")     # 60
print(f"Division (a / b): {a / b}")           # 3.75 (float)
print(f"Floor Division (a // b): {a // b}")   # 3 (integer)
print(f"Modulus (a % b): {a % b}")            # 3 (remainder)
print(f"Exponentiation (a ** b): {a ** b}")   # 50625 (15^4)
```

## 2. Comparison (Relational) Operators

Used to compare values. Returns `True` or `False`.

```python
x, y = 10, 20

print(f"x = {x}, y = {y}")
print(f"Equal (x == y): {x == y}")                # False
print(f"Not Equal (x != y): {x != y}")            # True
print(f"Greater Than (x > y): {x > y}")           # False
print(f"Less Than (x < y): {x < y}")              # True
print(f"Greater or Equal (x >= y): {x >= y}")     # False
print(f"Less or Equal (x <= y): {x <= y}")        # True

# Chained comparisons
z = 15
print(f"\nChained: {x} < {z} < {y} = {x < z < y}")  # True
```

## 3. Logical Operators

Used to combine conditional statements.

```python
a, b = True, False

print(f"a = {a}, b = {b}")
print(f"AND (a and b): {a and b}")    # False - both must be True
print(f"OR (a or b): {a or b}")       # True - at least one True
print(f"NOT (not a): {not a}")        # False - inverts value
print(f"NOT (not b): {not b}")        # True

# Practical example
age = 25
has_license = True
print(f"\nCan drive: {age >= 18 and has_license}")  # True

# Short-circuit evaluation
print(f"\nShort-circuit AND: {False and print('Not executed')}")
print(f"Short-circuit OR: {True or print('Not executed')}")
```

## 4. Assignment Operators

Used to assign values to variables.

```python
# Basic assignment
x = 10
print(f"x = 10 -> {x}")

# Add and assign
x += 5  # x = x + 5
print(f"x += 5 -> {x}")

# Subtract and assign
x -= 3  # x = x - 3
print(f"x -= 3 -> {x}")

# Multiply and assign
x *= 2  # x = x * 2
print(f"x *= 2 -> {x}")

# Divide and assign
x /= 4  # x = x / 4
print(f"x /= 4 -> {x}")

# Floor divide and assign
x = 17
x //= 3  # x = x // 3
print(f"x //= 3 -> {x}")

# Modulus and assign
x = 17
x %= 5  # x = x % 5
print(f"x %= 5 -> {x}")

# Exponent and assign
x = 2
x **= 4  # x = x ** 4
print(f"x **= 4 -> {x}")

# Walrus operator (Python 3.8+)
# Assigns value as part of expression
if (n := len([1, 2, 3])) > 2:
    print(f"\nWalrus operator: n = {n}")
```

## 5. Bitwise Operators

Used for bit-level operations on integers.

```python
a, b = 5, 3  # Binary: a = 0101, b = 0011

print(f"a = {a} (binary: {bin(a)})")
print(f"b = {b} (binary: {bin(b)})")
print()

# AND - 1 if both bits are 1
print(f"a & b (AND): {a & b} (binary: {bin(a & b)})")

# OR - 1 if at least one bit is 1
print(f"a | b (OR): {a | b} (binary: {bin(a | b)})")

# XOR - 1 if bits are different
print(f"a ^ b (XOR): {a ^ b} (binary: {bin(a ^ b)})")

# NOT - inverts all bits
print(f"~a (NOT): {~a} (binary: {bin(~a)})")

# Left Shift - shifts bits left, fills with 0
print(f"a << 1 (Left Shift): {a << 1} (binary: {bin(a << 1)})")

# Right Shift - shifts bits right
print(f"a >> 1 (Right Shift): {a >> 1} (binary: {bin(a >> 1)})")
```

## 6. Identity Operators

Used to compare object identity (memory location).

```python
# 'is' checks if two variables point to the same object
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a = {a}, id(a) = {id(a)}")
print(f"b = {b}, id(b) = {id(b)}")
print(f"c = a, id(c) = {id(c)}")
print()

print(f"a == b: {a == b}")      # True (same value)
print(f"a is b: {a is b}")      # False (different objects)
print(f"a is c: {a is c}")      # True (same object)
print(f"a is not b: {a is not b}")  # True

# Small integers are cached (-5 to 256)
x = 100
y = 100
print(f"\n100 is 100: {x is y}")  # True (cached)

x = 1000
y = 1000
print(f"1000 is 1000: {x is y}")  # May be False (not cached)
```

## 7. Membership Operators

Used to check if a value exists in a sequence.

```python
# in - returns True if value is found
# not in - returns True if value is NOT found

# With lists
fruits = ["apple", "banana", "cherry"]
print(f"'apple' in fruits: {'apple' in fruits}")          # True
print(f"'mango' in fruits: {'mango' in fruits}")          # False
print(f"'mango' not in fruits: {'mango' not in fruits}")  # True

# With strings
text = "Hello, World!"
print(f"\n'World' in text: {'World' in text}")            # True
print(f"'world' in text: {'world' in text}")              # False (case-sensitive)

# With dictionaries (checks keys)
person = {"name": "Alice", "age": 25}
print(f"\n'name' in person: {'name' in person}")          # True
print(f"'Alice' in person: {'Alice' in person}")          # False (not a key)
print(f"'Alice' in person.values(): {'Alice' in person.values()}")  # True

# With tuples
numbers = (1, 2, 3, 4, 5)
print(f"\n3 in numbers: {3 in numbers}")                  # True

# With sets
colors = {"red", "green", "blue"}
print(f"'red' in colors: {'red' in colors}")              # True
```

## 8. Operator Precedence

Order in which operators are evaluated (highest to lowest).

```python
# Precedence (highest to lowest):
# 1. () - Parentheses
# 2. ** - Exponentiation
# 3. +x, -x, ~x - Unary plus, minus, NOT
# 4. *, /, //, % - Multiplication, Division, Floor Div, Modulus
# 5. +, - - Addition, Subtraction
# 6. <<, >> - Bitwise shifts
# 7. & - Bitwise AND
# 8. ^ - Bitwise XOR
# 9. | - Bitwise OR
# 10. ==, !=, <, <=, >, >=, is, is not, in, not in - Comparisons
# 11. not - Logical NOT
# 12. and - Logical AND
# 13. or - Logical OR

# Examples
print(f"2 + 3 * 4 = {2 + 3 * 4}")            # 14 (not 20)
print(f"(2 + 3) * 4 = {(2 + 3) * 4}")        # 20
print(f"2 ** 3 ** 2 = {2 ** 3 ** 2}")        # 512 (right-to-left: 2^(3^2))
print(f"(2 ** 3) ** 2 = {(2 ** 3) ** 2}")    # 64
print(f"10 - 3 - 2 = {10 - 3 - 2}")          # 5 (left-to-right)
print(f"True or False and False = {True or False and False}")  # True (and before or)
```
