# Python File Handling Cheatsheet

Reading, writing, and manipulating files in Python.

## 1. Opening Files

File modes:
- `'r'` - Read (default)
- `'w'` - Write (overwrites)
- `'a'` - Append
- `'x'` - Create (fails if exists)
- `'r+'` - Read and write
- `'b'` - Binary mode
- `'t'` - Text mode (default)

```python
# Basic file opening
# file = open("filename.txt", "r")
# content = file.read()
# file.close()  # Important!

# Using 'with' statement (recommended - auto-closes)
# with open("filename.txt", "r") as file:
#     content = file.read()

# Create a sample file for demonstration
sample_content = """Line 1: Hello, World!
Line 2: Python is great!
Line 3: File handling is easy.
Line 4: This is the last line."""

with open("sample.txt", "w") as f:
    f.write(sample_content)
print("Sample file created!")
```

## 2. Reading Files

```python
# Read entire file
with open("sample.txt", "r") as file:
    content = file.read()
    print("Full content:")
    print(content)
```

```python
# Read specific number of characters
with open("sample.txt", "r") as file:
    first_20 = file.read(20)
    print(f"First 20 chars: '{first_20}'")
    
    # Continue reading
    next_20 = file.read(20)
    print(f"Next 20 chars: '{next_20}'")
```

```python
# Read line by line
with open("sample.txt", "r") as file:
    line1 = file.readline()
    line2 = file.readline()
    print(f"Line 1: {line1.strip()}")
    print(f"Line 2: {line2.strip()}")
```

```python
# Read all lines into list
with open("sample.txt", "r") as file:
    lines = file.readlines()
    print(f"Number of lines: {len(lines)}")
    for i, line in enumerate(lines):
        print(f"  {i}: {line.strip()}")
```

```python
# Iterate over file (memory efficient for large files)
with open("sample.txt", "r") as file:
    for line_num, line in enumerate(file, 1):
        print(f"{line_num}: {line.strip()}")
```

## 3. Writing Files

```python
# Write mode (overwrites existing content)
with open("output.txt", "w") as file:
    file.write("Hello, World!\n")
    file.write("This is a new file.\n")

with open("output.txt", "r") as file:
    print(file.read())
```

```python
# Write multiple lines
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]

with open("output.txt", "w") as file:
    file.writelines(lines)

with open("output.txt", "r") as file:
    print(file.read())
```

```python
# Append mode (adds to existing content)
with open("output.txt", "a") as file:
    file.write("Line 4 - Appended\n")
    file.write("Line 5 - Also appended\n")

with open("output.txt", "r") as file:
    print(file.read())
```

```python
# Create mode (fails if file exists)
import os

if os.path.exists("new_file.txt"):
    os.remove("new_file.txt")

with open("new_file.txt", "x") as file:
    file.write("This is a new file!\n")

print("New file created!")
```

## 4. File Position

```python
with open("sample.txt", "r") as file:
    # Get current position
    print(f"Initial position: {file.tell()}")
    
    # Read some content
    content = file.read(20)
    print(f"After reading 20 chars: {file.tell()}")
    
    # Seek to beginning
    file.seek(0)
    print(f"After seek(0): {file.tell()}")
    
    # Seek to specific position
    file.seek(10)
    print(f"After seek(10): {file.tell()}")
    print(f"Content from position 10: {file.read(15)}")
```

## 5. Binary Files

```python
# Write binary file
data = bytes([0, 1, 2, 3, 4, 5, 255])

with open("binary_file.bin", "wb") as file:
    file.write(data)

# Read binary file
with open("binary_file.bin", "rb") as file:
    content = file.read()
    print(f"Binary content: {content}")
    print(f"As list: {list(content)}")
```

```python
# Copy binary file (e.g., image)
def copy_file(source, destination):
    with open(source, "rb") as src:
        with open(destination, "wb") as dst:
            dst.write(src.read())

# For large files, read in chunks
def copy_large_file(source, destination, chunk_size=1024):
    with open(source, "rb") as src:
        with open(destination, "wb") as dst:
            while True:
                chunk = src.read(chunk_size)
                if not chunk:
                    break
                dst.write(chunk)
```

## 6. Working with Paths (os and pathlib)

```python
import os

# Current directory
print(f"Current dir: {os.getcwd()}")

# List files
print(f"\nFiles in current dir: {os.listdir('.')}")

# Check if path exists
print(f"\nsample.txt exists: {os.path.exists('sample.txt')}")
print(f"Is file: {os.path.isfile('sample.txt')}")
print(f"Is directory: {os.path.isdir('sample.txt')}")

# File info
print(f"\nFile size: {os.path.getsize('sample.txt')} bytes")

# Path manipulation
path = "/home/user/documents/file.txt"
print(f"\nDirectory: {os.path.dirname(path)}")
print(f"Filename: {os.path.basename(path)}")
print(f"Split: {os.path.split(path)}")
print(f"Extension: {os.path.splitext(path)}")

# Join paths
new_path = os.path.join("folder", "subfolder", "file.txt")
print(f"Joined path: {new_path}")
```

```python
from pathlib import Path

# Create Path object
path = Path("sample.txt")

print(f"Exists: {path.exists()}")
print(f"Is file: {path.is_file()}")
print(f"Name: {path.name}")
print(f"Stem: {path.stem}")
print(f"Suffix: {path.suffix}")
print(f"Parent: {path.parent}")

# Read/write with pathlib
content = path.read_text()
print(f"\nContent:\n{content[:50]}...")

# Create new file
new_path = Path("pathlib_example.txt")
new_path.write_text("Hello from pathlib!")
print(f"\nCreated: {new_path.read_text()}")
```

```python
# Directory operations with pathlib
from pathlib import Path

# Current directory
cwd = Path.cwd()
print(f"Current: {cwd}")

# Home directory
home = Path.home()
print(f"Home: {home}")

# Iterate over files
print("\nText files:")
for txt_file in Path(".").glob("*.txt"):
    print(f"  {txt_file}")

# Recursive glob
print("\nAll .py files (recursive):")
for py_file in Path(".").rglob("*.py"):
    print(f"  {py_file}")
```

## 7. Working with CSV Files

```python
import csv

# Write CSV
data = [
    ["Name", "Age", "City"],
    ["Alice", 25, "NYC"],
    ["Bob", 30, "LA"],
    ["Charlie", 35, "Chicago"]
]

with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created!")
```

```python
# Read CSV
with open("data.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# CSV with dictionaries
data = [
    {"Name": "Alice", "Age": 25, "City": "NYC"},
    {"Name": "Bob", "Age": 30, "City": "LA"},
    {"Name": "Charlie", "Age": 35, "City": "Chicago"}
]

# Write with DictWriter
with open("data_dict.csv", "w", newline="") as file:
    fieldnames = ["Name", "Age", "City"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data)

# Read with DictReader
with open("data_dict.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(dict(row))
```

## 8. Working with JSON Files

```python
import json

# Write JSON
data = {
    "name": "Alice",
    "age": 25,
    "cities": ["NYC", "LA", "Chicago"],
    "active": True
}

with open("data.json", "w") as file:
    json.dump(data, file, indent=2)

print("JSON file created!")
```

```python
# Read JSON
with open("data.json", "r") as file:
    loaded_data = json.load(file)
    print(loaded_data)
    print(f"Name: {loaded_data['name']}")
```

```python
# JSON string operations
data = {"name": "Bob", "age": 30}

# Convert to JSON string
json_string = json.dumps(data)
print(f"JSON string: {json_string}")

# Parse JSON string
parsed = json.loads(json_string)
print(f"Parsed: {parsed}")
```

## 9. Exception Handling with Files

Common file-related exceptions:
- `FileNotFoundError` - File doesn't exist
- `PermissionError` - No permission to access
- `IsADirectoryError` - Expected file, got directory
- `FileExistsError` - File already exists (with 'x' mode)
- `IOError` / `OSError` - General I/O errors

```python
# Handling FileNotFoundError
try:
    with open("nonexistent_file.txt", "r") as file:
        content = file.read()
except FileNotFoundError:
    print("Error: File not found!")
except PermissionError:
    print("Error: No permission to read file!")
```

```python
# Handling FileExistsError with 'x' mode
try:
    # First create a file
    with open("test_exists.txt", "w") as f:
        f.write("test")
    
    # Try to create again with 'x' mode
    with open("test_exists.txt", "x") as file:
        file.write("This will fail!")
except FileExistsError:
    print("Error: File already exists!")
finally:
    # Cleanup
    import os
    if os.path.exists("test_exists.txt"):
        os.remove("test_exists.txt")
```

```python
# Handling IsADirectoryError
import os

try:
    # Try to open a directory as a file
    with open(".", "r") as file:
        content = file.read()
except IsADirectoryError:
    print("Error: Cannot read a directory as a file!")
```

## File Handling Summary

|------|-------------|

| `'a'` | Append |
| `'x'` | Create (fails if exists) |
| `'r+'` | Read and write |
| `'b'` | Binary mode |

| Method | Description |
|--------|-------------|
| `read()` | Read entire file |
| `read(n)` | Read n characters |
| `readline()` | Read one line |
| `readlines()` | Read all lines to list |
| `write()` | Write string |
| `writelines()` | Write list of strings |
| `seek(n)` | Move to position n |
| `tell()` | Get current position |
