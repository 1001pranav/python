# Python Regular Expressions Cheatsheet

Pattern matching and text manipulation with the `re` module.

## 1. Basic Patterns

```python
import re

# Match literal text
text = "Hello, World!"
pattern = r"World"

match = re.search(pattern, text)
if match:
    print(f"Found: {match.group()}")
    print(f"Position: {match.start()} to {match.end()}")
```

```python
# Special characters (metacharacters)
# .  - Any character except newline
# ^  - Start of string
# $  - End of string
# *  - 0 or more
# +  - 1 or more
# ?  - 0 or 1
# {} - Exact number
# [] - Character class
# |  - Or
# () - Group
# \  - Escape

# . matches any character
print(re.findall(r"h.t", "hat hot hut hit"))
```

## 2. Character Classes

```python
text = "The price is $100 and 50% off"

# [abc] - Match a, b, or c
print(f"[aeiou]: {re.findall(r'[aeiou]', text)}")

# [^abc] - Match anything except a, b, c
print(f"[^aeiou]: {re.findall(r'[^aeiou ]', text)[:10]}...")

# [a-z] - Range
print(f"[a-z]: {re.findall(r'[a-z]+', text)}")

# [0-9] - Digits
print(f"[0-9]+: {re.findall(r'[0-9]+', text)}")
```

```python
# Predefined character classes
text = "Hello World 123 test_var"

# \d - Digit [0-9]
print(f"\\d+: {re.findall(r'\d+', text)}")

# \D - Non-digit [^0-9]
print(f"\\D+: {re.findall(r'\D+', text)}")

# \w - Word character [a-zA-Z0-9_]
print(f"\\w+: {re.findall(r'\w+', text)}")

# \W - Non-word character
print(f"\\W+: {re.findall(r'\W+', text)}")

# \s - Whitespace [ \t\n\r\f\v]
print(f"\\s+: {re.findall(r'\s+', text)}")

# \S - Non-whitespace
print(f"\\S+: {re.findall(r'\S+', text)}")
```

## 3. Quantifiers

```python
text = "goood good gd god gooood"

# * - 0 or more
print(f"go*d: {re.findall(r'go*d', text)}")

# + - 1 or more
print(f"go+d: {re.findall(r'go+d', text)}")

# ? - 0 or 1
print(f"go?d: {re.findall(r'go?d', text)}")

# {n} - Exactly n
print(f"go{{2}}d: {re.findall(r'go{2}d', text)}")

# {n,m} - n to m
print(f"go{{2,3}}d: {re.findall(r'go{2,3}d', text)}")

# {n,} - n or more
print(f"go{{2,}}d: {re.findall(r'go{2,}d', text)}")
```

```python
# Greedy vs Non-greedy
text = "<tag>content</tag>"

# Greedy (default) - matches as much as possible
print(f"Greedy <.*>: {re.findall(r'<.*>', text)}")

# Non-greedy (?) - matches as little as possible
print(f"Non-greedy <.*?>: {re.findall(r'<.*?>', text)}")
```

## 4. Anchors and Boundaries

```python
text = "hello world\nhello python"

# ^ - Start of string (or line with MULTILINE)
print(f"^hello: {re.findall(r'^hello', text)}")
print(f"^hello (MULTILINE): {re.findall(r'^hello', text, re.MULTILINE)}")

# $ - End of string (or line with MULTILINE)
print(f"world$: {re.findall(r'world$', text)}")
print(f"python$: {re.findall(r'python$', text)}")
```

```python
# \b - Word boundary
text = "cat category catfish scat"

print(f"cat: {re.findall(r'cat', text)}")
print(f"\\bcat\\b: {re.findall(r'\bcat\b', text)}")
print(f"\\bcat: {re.findall(r'\bcat', text)}")
print(f"cat\\b: {re.findall(r'cat\b', text)}")
```

## 5. Groups and Capturing

```python
# Basic groups
text = "John Smith, Jane Doe, Bob Wilson"

# Capture first and last name
pattern = r"(\w+) (\w+)"
matches = re.findall(pattern, text)
print(f"Names: {matches}")

# Access groups
match = re.search(pattern, text)
if match:
    print(f"Full match: {match.group(0)}")
    print(f"First name: {match.group(1)}")
    print(f"Last name: {match.group(2)}")
    print(f"All groups: {match.groups()}")
```

```python
# Named groups
text = "Date: 2024-01-15"

pattern = r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})"
match = re.search(pattern, text)

if match:
    print(f"Year: {match.group('year')}")
    print(f"Month: {match.group('month')}")
    print(f"Day: {match.group('day')}")
    print(f"Dict: {match.groupdict()}")
```

```python
# Non-capturing group (?:...)
text = "cat cats dog dogs"

# Capturing
print(f"Capturing: {re.findall(r'(cat|dog)s?', text)}")

# Non-capturing
print(f"Non-capturing: {re.findall(r'(?:cat|dog)s?', text)}")
```

```python
# Backreferences
text = "hello hello world world"

# Find repeated words
pattern = r"\b(\w+)\s+\1\b"
print(f"Repeated words: {re.findall(pattern, text)}")
```

## 6. re Module Functions

```python
text = "The rain in Spain falls mainly on the plain"

# re.search() - Find first match
match = re.search(r"\b\w+ain\b", text)
print(f"search: {match.group() if match else None}")

# re.match() - Match at beginning only
match = re.match(r"The", text)
print(f"match 'The': {match.group() if match else None}")
match = re.match(r"rain", text)
print(f"match 'rain': {match}")

# re.fullmatch() - Match entire string
match = re.fullmatch(r"\w+", "hello")
print(f"fullmatch: {match.group() if match else None}")

# re.findall() - Find all matches
print(f"findall: {re.findall(r'\b\w+ain\b', text)}")

# re.finditer() - Iterator of match objects
for match in re.finditer(r"\b\w+ain\b", text):
    print(f"finditer: {match.group()} at {match.start()}")
```

```python
# re.sub() - Replace
text = "Hello World"

# Simple replacement
result = re.sub(r"World", "Python", text)
print(f"sub: {result}")

# Replace with function
def upper_repl(match):
    return match.group().upper()

text = "hello world"
result = re.sub(r"\b\w", upper_repl, text)
print(f"sub with func: {result}")

# Replace with backreference
text = "hello world"
result = re.sub(r"(\w+) (\w+)", r"\2 \1", text)
print(f"sub with backref: {result}")

# re.subn() - Replace and count
result, count = re.subn(r"\d", "X", "a1b2c3")
print(f"subn: {result}, count: {count}")
```

```python
# re.split() - Split by pattern
text = "one1two2three3four"
result = re.split(r"\d", text)
print(f"split: {result}")

# Keep delimiters with capturing group
result = re.split(r"(\d)", text)
print(f"split with delim: {result}")

# Max splits
result = re.split(r"\d", text, maxsplit=2)
print(f"split maxsplit=2: {result}")
```

```python
# Compiled patterns
pattern = re.compile(r"\b\w{4}\b")  # 4-letter words

text1 = "This is a test"
text2 = "More text here"

print(f"Text 1: {pattern.findall(text1)}")
print(f"Text 2: {pattern.findall(text2)}")
```

## 7. Flags

```python
text = "Hello\nHELLO\nhello"

# re.IGNORECASE or re.I - Case insensitive
print(f"IGNORECASE: {re.findall(r'hello', text, re.IGNORECASE)}")

# re.MULTILINE or re.M - ^ and $ match line boundaries
print(f"MULTILINE ^hello: {re.findall(r'^hello', text, re.MULTILINE | re.IGNORECASE)}")

# re.DOTALL or re.S - . matches newline too
text2 = "Hello\nWorld"
print(f"Without DOTALL: {re.findall(r'Hello.World', text2)}")
print(f"With DOTALL: {re.findall(r'Hello.World', text2, re.DOTALL)}")
```

```python
# re.VERBOSE or re.X - Allow comments and whitespace
pattern = re.compile(r"""
    \b          # Word boundary
    \d{3}       # Area code (3 digits)
    [-.]?       # Optional separator
    \d{3}       # First 3 digits
    [-.]?       # Optional separator
    \d{4}       # Last 4 digits
    \b          # Word boundary
""", re.VERBOSE)

text = "Call 123-456-7890 or 123.456.7890"
print(f"Phone numbers: {pattern.findall(text)}")
```

## 8. Common Patterns

```python
# Email validation
email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b"
text = "Contact us at info@example.com or support@test.org"
print(f"Emails: {re.findall(email_pattern, text)}")
```

```python
# URL pattern
url_pattern = r"https?://[\w.-]+(?:/[\w./-]*)?"
text = "Visit https://example.com/page or http://test.org"
print(f"URLs: {re.findall(url_pattern, text)}")
```

```python
# Phone number
phone_pattern = r"\b\d{3}[-.]?\d{3}[-.]?\d{4}\b"
text = "Call 123-456-7890 or 123.456.7890 or 1234567890"
print(f"Phones: {re.findall(phone_pattern, text)}")
```

```python
# IP address
ip_pattern = r"\b(?:\d{1,3}\.){3}\d{1,3}\b"
text = "Server at 192.168.1.1 or 10.0.0.1"
print(f"IPs: {re.findall(ip_pattern, text)}")
```

```python
# Date (various formats)
date_pattern = r"\b\d{1,2}[-/]\d{1,2}[-/]\d{2,4}\b"
text = "Dates: 01/15/2024, 1-15-24, 15/01/2024"
print(f"Dates: {re.findall(date_pattern, text)}")
```

## Regex Quick Reference

| Pattern | Meaning |
|---------|---------|
| `.` | Any character |
| `\d` | Digit [0-9] |
| `\w` | Word char [a-zA-Z0-9_] |
| `\s` | Whitespace |
| `^` | Start of string |
| `$` | End of string |
| `\b` | Word boundary |
| `*` | 0 or more |
| `+` | 1 or more |
| `?` | 0 or 1 |
| `{n}` | Exactly n |
| `{n,m}` | n to m |
| `[abc]` | a, b, or c |
| `[^abc]` | Not a, b, c |
| `(...)` | Capture group |
| `(?:...)` | Non-capturing |
| `(?P<name>...)` | Named group |
| `\1` | Backreference |
