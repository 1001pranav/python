## 2. Common Built-in Modules

````python
# os - Operating system interface
import os

print(f"Current directory: {os.getcwd()}")
print(f"Environment variable: {os.environ.get('HOME', 'Not set')}")
# os.listdir('.')  # List directory contents
# os.makedirs('path/to/dir')  # Create directories
# os.remove('file.txt')  # Delete file
````

````python
# sys - System-specific parameters
import sys

print(f"Python version: {sys.version}")
print(f"Platform: {sys.platform}")
print(f"Path: {sys.path[:2]}...")  # Module search path
````

````python
# datetime - Date and time
from datetime import datetime, date, timedelta

now = datetime.now()
print(f"Now: {now}")
print(f"Formatted: {now.strftime('%Y-%m-%d %H:%M:%S')}")

# Timedelta
tomorrow = date.today() + timedelta(days=1)
print(f"Tomorrow: {tomorrow}")
````

````python
# random - Random number generation
import random

print(f"Random int [1-10]: {random.randint(1, 10)}")
print(f"Random choice: {random.choice(['a', 'b', 'c'])}")
print(f"Random float [0-1]: {random.random():.4f}")
````

````python
# json - JSON encoding/decoding
import json

data = {"name": "Alice", "age": 25}
json_str = json.dumps(data, indent=2)
print(f"JSON string:\n{json_str}")

parsed = json.loads(json_str)
print(f"\nParsed: {parsed}")
````

````python
# collections - Specialized containers
from collections import Counter, defaultdict, namedtuple, deque

# Counter
words = ['apple', 'banana', 'apple', 'cherry', 'apple']
counter = Counter(words)
print(f"Counter: {counter}")
print(f"Most common: {counter.most_common(2)}")

# defaultdict
dd = defaultdict(list)
dd['fruits'].append('apple')
dd['fruits'].append('banana')
print(f"\ndefaultdict: {dict(dd)}")

# namedtuple
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)
print(f"\nNamedtuple: {p}, x={p.x}, y={p.y}")

# deque
dq = deque([1, 2, 3])
dq.appendleft(0)
dq.append(4)
print(f"\nDeque: {dq}")
````

````python
# itertools - Iterator tools
import itertools

# count - infinite counter
counter = itertools.count(start=1, step=2)
print(f"Count: {[next(counter) for _ in range(5)]}")

# cycle - infinite cycle
cycler = itertools.cycle(['A', 'B', 'C'])
print(f"Cycle: {[next(cycler) for _ in range(6)]}")

# combinations
print(f"Combinations: {list(itertools.combinations('ABC', 2))}")

# permutations
print(f"Permutations: {list(itertools.permutations('AB', 2))}")

# product (cartesian product)
print(f"Product: {list(itertools.product('AB', '12'))}")
````

````python
# functools - Higher-order functions
from functools import reduce, lru_cache, partial

# reduce
numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
print(f"Product: {product}")

# lru_cache (memoization)
@lru_cache(maxsize=None)
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(f"Fibonacci(30): {fibonacci(30)}")

# partial (fix some arguments)
def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
cube = partial(power, exponent=3)
print(f"Square(5): {square(5)}, Cube(5): {cube(5)}")
````

## 3. Creating Modules

````python
# A module is just a .py file
# Example: mymodule.py

module_code = '''
"""This is my custom module."""

# Module-level variable
VERSION = "1.0.0"

def greet(name):
    """Greet someone."""
    return f"Hello, {name}!"

def add(a, b):
    """Add two numbers."""
    return a + b

class Calculator:
    """Simple calculator class."""
    def multiply(self, a, b):
        return a * b

# This runs only when module is executed directly
if __name__ == "__main__":
    print("Running as main program")
    print(greet("World"))
'''

# Write the module file
with open('mymodule.py', 'w') as f:
    f.write(module_code)

print("Module created!")
````

````python
# Import and use custom module
import mymodule

print(f"Version: {mymodule.VERSION}")
print(mymodule.greet("Alice"))
print(f"Add: {mymodule.add(3, 5)}")

calc = mymodule.Calculator()
print(f"Multiply: {calc.multiply(4, 5)}")
````

````python
# __name__ and __main__
# When a module is imported: __name__ = 'module_name'
# When a module is run directly: __name__ = '__main__'

print(f"Current module __name__: {__name__}")
````

## 4. Creating Packages

A package is a directory containing modules and an `__init__.py` file.

```
mypackage/
    __init__.py
    module1.py
    module2.py
    subpackage/
        __init__.py
        module3.py
```

````python
import os

# Create package structure
os.makedirs('mypackage/subpackage', exist_ok=True)

# __init__.py for main package
init_code = '''
"""My Package."""
from .module1 import func1
from .module2 import func2

__all__ = ['func1', 'func2', 'module1', 'module2']
__version__ = '1.0.0'
'''

# module1.py
module1_code = '''
def func1():
    return "Hello from module1"
'''

# module2.py
module2_code = '''
def func2():
    return "Hello from module2"
'''

# subpackage/__init__.py
sub_init_code = '''
from .module3 import func3
'''

# subpackage/module3.py
module3_code = '''
def func3():
    return "Hello from subpackage module3"
'''

# Write files
with open('mypackage/__init__.py', 'w') as f:
    f.write(init_code)
with open('mypackage/module1.py', 'w') as f:
    f.write(module1_code)
with open('mypackage/module2.py', 'w') as f:
    f.write(module2_code)
with open('mypackage/subpackage/__init__.py', 'w') as f:
    f.write(sub_init_code)
with open('mypackage/subpackage/module3.py', 'w') as f:
    f.write(module3_code)

print("Package created!")
````

````python
# Using the package
import mypackage
from mypackage import module1
from mypackage.subpackage import module3

print(f"Package version: {mypackage.__version__}")
print(mypackage.func1())  # From __init__.py import
print(module1.func1())
print(module3.func3())
````
