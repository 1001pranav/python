## Task 1 Safe Division Utility

```python
def division(numerator, denominator):
    try:
        if type(numerator) is str or type(denominator) is str:
            raise ValueError("Both or any one number should not be string")
        elif denominator == 0:
            raise ZeroDivisionError("Denominator Should not be zero")
        else: 
            print(numerator/denominator)
    except ValueError as e:
        print(e)
    except ZeroDivisionError as e:
        print(e)
    finally:
        print("Operation Completed")
```

```python
division(25, 30)
```

```python
division('1', '9')
```

```python
division(45, 0)
```

## Task 2 Bill Calculator with Error Handling

```python
def bill_calculator(price_lists):
    total = 0
    for price in price_lists:
        try:
            if type(price) is str:
                raise TypeError("Value is not a number")
            elif price < 0:
                raise ValueError("Negative Price not allowed")
            else: total += price
            print("Total: ", total)
        except TypeError as E:
            print(E)
            continue
        except ValueError as E:
            print(E)
            continue
```

```python
bill_calculator([120, 350, 110, -100, '15004', 160])
```

## Task 3 Custom Age Validation

```python
def custom_age_validation(age: int):
    if type(age) is str:
        raise TypeError("Age must be string")
    elif age < 0 or age > 121:
        raise ValueError("Age Must be within 0 or 120")
    else:
        print("Age entered is ", age)
```

```python
try:
    age_string = input("Enter Age")
    custom_age_validation(int(age_string))
except TypeError as E:
    print(E)
except ValueError as E:
    print(E)
```

## Task 4 File Reader Exception

```python
import os
```

```python
file_name = input('Enter a filename which is present in assets folder')
file_path = os.path.join('assets', file_name)
```

```python
try:
    with open(file_path, 'r') as file_class:
        text = file_class.read()
        print(text)
except FileNotFoundError as _:
    print(f"File You are looking {file_name} did not found")
except PermissionError as _:
    print(f"Don't have relevant permission to read the file {file_name}")
finally:
    print("File Operations attempted")
```
