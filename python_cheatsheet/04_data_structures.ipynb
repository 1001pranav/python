# Python Data Structures Cheatsheet

Python has four built-in data structures: Lists, Tuples, Sets, and Dictionaries.

## 1. Lists

Lists are ordered, mutable collections that allow duplicate values.

```python
# Creating lists
empty_list = []
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True, None]
nested = [[1, 2], [3, 4], [5, 6]]
from_range = list(range(1, 6))

print(f"Numbers: {numbers}")
print(f"Mixed: {mixed}")
print(f"Nested: {nested}")
print(f"From range: {from_range}")
```

```python
# Accessing elements
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

print(f"First: {fruits[0]}")
print(f"Last: {fruits[-1]}")
print(f"Slice [1:3]: {fruits[1:3]}")
print(f"Reverse: {fruits[::-1]}")

# Nested list access
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(f"\nMatrix[1][2]: {matrix[1][2]}")
```

```python
# Modifying lists
colors = ["red", "green", "blue"]
print(f"Original: {colors}")

# Change element
colors[1] = "yellow"
print(f"After change: {colors}")

# Append - add to end
colors.append("purple")
print(f"After append: {colors}")

# Insert at index
colors.insert(1, "orange")
print(f"After insert: {colors}")

# Extend - add multiple elements
colors.extend(["pink", "brown"])
print(f"After extend: {colors}")

# Remove by value
colors.remove("yellow")
print(f"After remove: {colors}")

# Pop - remove and return
last = colors.pop()
print(f"Popped: {last}, List: {colors}")

# Pop at index
second = colors.pop(1)
print(f"Popped index 1: {second}, List: {colors}")

# Clear all elements
temp = [1, 2, 3]
temp.clear()
print(f"After clear: {temp}")
```

```python
# List methods
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]

print(f"Original: {numbers}")
print(f"Count of 5: {numbers.count(5)}")
print(f"Index of 9: {numbers.index(9)}")

# Sort in place
numbers.sort()
print(f"Sorted: {numbers}")

numbers.sort(reverse=True)
print(f"Sorted desc: {numbers}")

# Reverse in place
numbers.reverse()
print(f"Reversed: {numbers}")

# Copy
copy1 = numbers.copy()
copy2 = numbers[:]  # slice copy
copy3 = list(numbers)
print(f"\nCopy: {copy1}")
```

```python
# List unpacking
colors = ["red", "green", "blue", "yellow", "purple"]

first, second, *rest = colors
print(f"first: {first}, second: {second}, rest: {rest}")

first, *middle, last = colors
print(f"first: {first}, middle: {middle}, last: {last}")

# List operations
a = [1, 2, 3]
b = [4, 5, 6]
print(f"\nConcatenate: {a + b}")
print(f"Repeat: {a * 3}")
print(f"Length: {len(a)}")
print(f"Min: {min(a)}, Max: {max(a)}, Sum: {sum(a)}")
print(f"3 in a: {3 in a}")
```

## 2. Tuples

Tuples are ordered, immutable collections that allow duplicate values.

```python
# Creating tuples
empty_tuple = ()
single = (1,)  # Note: comma is required for single element
not_tuple = (1)  # This is just an integer
numbers = (1, 2, 3, 4, 5)
mixed = (1, "hello", 3.14)
from_list = tuple([1, 2, 3])

print(f"Single element tuple: {single}, type: {type(single)}")
print(f"Not a tuple: {not_tuple}, type: {type(not_tuple)}")
print(f"Numbers: {numbers}")
```

```python
# Accessing tuples (same as lists)
coords = (10, 20, 30, 40, 50)

print(f"First: {coords[0]}")
print(f"Last: {coords[-1]}")
print(f"Slice [1:4]: {coords[1:4]}")

# Tuple unpacking
point = (3, 4, 5)
x, y, z = point
print(f"\nx: {x}, y: {y}, z: {z}")

# Swap using tuples
a, b = 10, 20
a, b = b, a
print(f"Swapped: a={a}, b={b}")
```

```python
# Tuple methods and operations
numbers = (1, 2, 3, 2, 4, 2, 5)

print(f"count(2): {numbers.count(2)}")
print(f"index(3): {numbers.index(3)}")
print(f"Length: {len(numbers)}")
print(f"Min: {min(numbers)}, Max: {max(numbers)}")

# Concatenate
t1 = (1, 2)
t2 = (3, 4)
print(f"\nConcatenate: {t1 + t2}")
print(f"Repeat: {t1 * 3}")

# Convert to list to modify, then back to tuple
temp = list(numbers)
temp.append(6)
numbers = tuple(temp)
print(f"Modified: {numbers}")
```

## 3. Sets

Sets are unordered collections of unique elements.

```python
# Creating sets
empty_set = set()  # Note: {} creates an empty dict
numbers = {1, 2, 3, 4, 5}
from_list = set([1, 2, 2, 3, 3, 3])  # Duplicates removed
from_string = set("hello")  # Each character

print(f"Numbers: {numbers}")
print(f"From list (duplicates removed): {from_list}")
print(f"From string: {from_string}")
```

```python
# Modifying sets
fruits = {"apple", "banana", "cherry"}
print(f"Original: {fruits}")

# Add single element
fruits.add("date")
print(f"After add: {fruits}")

# Update - add multiple elements
fruits.update(["elderberry", "fig"])
print(f"After update: {fruits}")

# Remove (raises error if not found)
fruits.remove("banana")
print(f"After remove: {fruits}")

# Discard (no error if not found)
fruits.discard("grape")  # No error
print(f"After discard: {fruits}")

# Pop - remove and return arbitrary element
popped = fruits.pop()
print(f"Popped: {popped}")

# Clear
temp = {1, 2, 3}
temp.clear()
print(f"After clear: {temp}")
```

```python
# Set operations
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(f"Set A: {a}")
print(f"Set B: {b}")

# Union - all elements from both
print(f"\nUnion (a | b): {a | b}")
print(f"Union (a.union(b)): {a.union(b)}")

# Intersection - common elements
print(f"\nIntersection (a & b): {a & b}")
print(f"Intersection (a.intersection(b)): {a.intersection(b)}")

# Difference - elements in a but not in b
print(f"\nDifference (a - b): {a - b}")
print(f"Difference (a.difference(b)): {a.difference(b)}")

# Symmetric difference - elements in either but not both
print(f"\nSymmetric diff (a ^ b): {a ^ b}")
print(f"Symmetric diff: {a.symmetric_difference(b)}")
```

```python
# Set comparisons
a = {1, 2, 3}
b = {1, 2, 3, 4, 5}
c = {1, 2, 3}

print(f"a: {a}, b: {b}, c: {c}")
print(f"a == c: {a == c}")
print(f"a.issubset(b): {a.issubset(b)}")
print(f"b.issuperset(a): {b.issuperset(a)}")
print(f"a.isdisjoint({4,5}): {a.isdisjoint({4,5})}")

# Frozen set - immutable set
frozen = frozenset([1, 2, 3])
print(f"\nFrozenset: {frozen}")
# frozen.add(4)  # This would raise an error
```

## 4. Dictionaries

Dictionaries are unordered collections of key-value pairs.

```python
# Creating dictionaries
empty_dict = {}
person = {"name": "Alice", "age": 25, "city": "NYC"}
from_tuples = dict([("a", 1), ("b", 2)])
from_keys = dict.fromkeys(["x", "y", "z"], 0)
using_constructor = dict(name="Bob", age=30)

print(f"Person: {person}")
print(f"From tuples: {from_tuples}")
print(f"From keys: {from_keys}")
print(f"Using constructor: {using_constructor}")
```

```python
# Accessing dictionaries
person = {"name": "Alice", "age": 25, "city": "NYC"}

# Direct access (raises KeyError if not found)
print(f"Name: {person['name']}")

# get() - returns None or default if not found
print(f"Age: {person.get('age')}")
print(f"Country: {person.get('country')}")
print(f"Country (default): {person.get('country', 'Unknown')}")

# Keys, values, items
print(f"\nKeys: {list(person.keys())}")
print(f"Values: {list(person.values())}")
print(f"Items: {list(person.items())}")
```

```python
# Modifying dictionaries
person = {"name": "Alice", "age": 25}
print(f"Original: {person}")

# Add/update single item
person["city"] = "NYC"
person["age"] = 26
print(f"After add/update: {person}")

# Update multiple items
person.update({"country": "USA", "age": 27})
print(f"After update: {person}")

# setdefault - add if not exists
person.setdefault("email", "alice@email.com")
person.setdefault("name", "Bob")  # Won't change, key exists
print(f"After setdefault: {person}")

# Remove items
del person["email"]
print(f"After del: {person}")

# pop - remove and return value
age = person.pop("age")
print(f"Popped age: {age}, Dict: {person}")

# popitem - remove and return last item
item = person.popitem()
print(f"Popped item: {item}")

# Clear
person.clear()
print(f"After clear: {person}")
```

```python
# Iterating dictionaries
person = {"name": "Alice", "age": 25, "city": "NYC"}

# Iterate keys
print("Keys:")
for key in person:
    print(f"  {key}")

# Iterate values
print("\nValues:")
for value in person.values():
    print(f"  {value}")

# Iterate key-value pairs
print("\nItems:")
for key, value in person.items():
    print(f"  {key}: {value}")
```

```python
# Nested dictionaries
users = {
    "user1": {"name": "Alice", "age": 25},
    "user2": {"name": "Bob", "age": 30},
    "user3": {"name": "Charlie", "age": 35}
}

print(f"User1 name: {users['user1']['name']}")

# Add nested key
users["user1"]["email"] = "alice@email.com"

# Iterate nested
for user_id, info in users.items():
    print(f"{user_id}: {info}")
```

```python
# Dictionary copy
original = {"a": 1, "b": [1, 2, 3]}

# Shallow copy
shallow = original.copy()
shallow["b"].append(4)  # Modifies original too!
print(f"Original: {original}")
print(f"Shallow: {shallow}")

# Deep copy
import copy
original = {"a": 1, "b": [1, 2, 3]}
deep = copy.deepcopy(original)
deep["b"].append(4)
print(f"\nOriginal: {original}")
print(f"Deep: {deep}")
```

## Data Structures Comparison Table

| Feature | List | Tuple | Set | Dictionary |
|---------|------|-------|-----|------------|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` | `{1, 2, 3}` | `{"a": 1}` |
| Ordered | Yes | Yes | No | Yes (3.7+) |
| Mutable | Yes | No | Yes | Yes |
| Duplicates | Yes | Yes | No | Keys: No |
| Indexing | Yes | Yes | No | By key |
| Use case | General collection | Fixed data | Unique items | Key-value pairs |
