# Python Control Flow Cheatsheet

Control flow statements determine the order in which code is executed.

## 1. If-Elif-Else Statements

```python
# Basic if statement
age = 18

if age >= 18:
    print("You are an adult")
```

```python
# If-else statement
temperature = 25

if temperature > 30:
    print("It's hot!")
else:
    print("It's not hot")
```

```python
# If-elif-else statement
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
```

```python
# Nested if statements
num = 15

if num > 0:
    if num % 2 == 0:
        print(f"{num} is positive and even")
    else:
        print(f"{num} is positive and odd")
else:
    print(f"{num} is not positive")
```

```python
# Ternary operator (conditional expression)
x = 10

# Syntax: value_if_true if condition else value_if_false
result = "Even" if x % 2 == 0 else "Odd"
print(f"{x} is {result}")

# Nested ternary
num = 0
sign = "positive" if num > 0 else "negative" if num < 0 else "zero"
print(f"{num} is {sign}")
```

```python
# Multiple conditions with and, or, not
age = 25
has_license = True
is_drunk = False

if age >= 18 and has_license and not is_drunk:
    print("You can drive")
else:
    print("You cannot drive")

# Checking multiple values
day = "Saturday"
if day in ["Saturday", "Sunday"]:
    print("It's the weekend!")
```

## 2. Match Statement (Python 3.10+)

Pattern matching - similar to switch/case in other languages.

```python
# Basic match statement
def get_day_type(day):
    match day:
        case "Monday" | "Tuesday" | "Wednesday" | "Thursday" | "Friday":
            return "Weekday"
        case "Saturday" | "Sunday":
            return "Weekend"
        case _:  # Default case (wildcard)
            return "Invalid day"

print(get_day_type("Monday"))
print(get_day_type("Saturday"))
print(get_day_type("Holiday"))
```

```python
# Match with patterns
def process_point(point):
    match point:
        case (0, 0):
            return "Origin"
        case (x, 0):
            return f"On X-axis at {x}"
        case (0, y):
            return f"On Y-axis at {y}"
        case (x, y):
            return f"Point at ({x}, {y})"
        case _:
            return "Invalid point"

print(process_point((0, 0)))
print(process_point((5, 0)))
print(process_point((3, 4)))
```

## 3. For Loops

```python
# Iterate over list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# Using range()
print("range(5):")
for i in range(5):  # 0 to 4
    print(i, end=" ")

print("\n\nrange(2, 8):")
for i in range(2, 8):  # 2 to 7
    print(i, end=" ")

print("\n\nrange(0, 10, 2):")
for i in range(0, 10, 2):  # 0, 2, 4, 6, 8
    print(i, end=" ")

print("\n\nrange(10, 0, -2):")
for i in range(10, 0, -2):  # 10, 8, 6, 4, 2
    print(i, end=" ")
```

```python
# Iterate over string
for char in "Python":
    print(char, end=" ")
```

```python
# Iterate with index using enumerate()
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")

# Start from different index
print("\nStarting from 1:")
for index, fruit in enumerate(fruits, start=1):
    print(f"{index}: {fruit}")
```

```python
# Iterate over dictionary
person = {"name": "Alice", "age": 25, "city": "NYC"}

print("Keys:")
for key in person:
    print(f"  {key}")

print("\nValues:")
for value in person.values():
    print(f"  {value}")

print("\nKey-Value pairs:")
for key, value in person.items():
    print(f"  {key}: {value}")
```

```python
# Iterate over multiple lists with zip()
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
cities = ["NYC", "LA", "Chicago"]

for name, age, city in zip(names, ages, cities):
    print(f"{name}, {age}, {city}")
```

```python
# Nested loops
for i in range(3):
    for j in range(3):
        print(f"({i}, {j})", end=" ")
    print()  # New line
```

```python
# For-else (else runs if loop completes without break)
numbers = [1, 3, 5, 7, 9]

for num in numbers:
    if num % 2 == 0:
        print(f"Found even number: {num}")
        break
else:
    print("No even number found")
```

## 4. While Loops

```python
# Basic while loop
count = 0
while count < 5:
    print(count, end=" ")
    count += 1
```

```python
# While with else
count = 0
while count < 3:
    print(f"Count: {count}")
    count += 1
else:
    print("Loop completed normally")
```

```python
# Infinite loop with break
count = 0
while True:
    print(count, end=" ")
    count += 1
    if count >= 5:
        break
```

## 5. Break, Continue, and Pass

```python
# break - exit the loop
print("Break example:")
for i in range(10):
    if i == 5:
        break
    print(i, end=" ")
print("\nLoop exited at 5")
```

```python
# continue - skip current iteration
print("Continue example (skip evens):")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i, end=" ")
```

```python
# pass - do nothing (placeholder)
for i in range(5):
    if i == 2:
        pass  # Placeholder for future code
    else:
        print(i, end=" ")

# Useful for empty functions/classes
def my_function():
    pass  # To be implemented later

class MyClass:
    pass
```

```python
# Break in nested loops (only breaks inner loop)
print("Break in nested loops:")
for i in range(3):
    for j in range(3):
        if j == 2:
            break
        print(f"({i},{j})", end=" ")
    print()  # New line

# To break outer loop, use flag or function
print("\nBreaking outer loop with flag:")
found = False
for i in range(3):
    for j in range(3):
        print(f"({i},{j})", end=" ")
        if i == 1 and j == 1:
            found = True
            break
    if found:
        break
```

## 6. Practical Examples

```python
# FizzBuzz
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz", end=" ")
    elif i % 3 == 0:
        print("Fizz", end=" ")
    elif i % 5 == 0:
        print("Buzz", end=" ")
    else:
        print(i, end=" ")
```

```python
# Find prime numbers
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [n for n in range(2, 30) if is_prime(n)]
print(f"Primes under 30: {primes}")
```

```python
# Factorial using while loop
def factorial(n):
    result = 1
    while n > 1:
        result *= n
        n -= 1
    return result

print(f"5! = {factorial(5)}")
```

## Control Flow Summary

| Statement | Purpose | Example |
|-----------|---------|----------|
| `if` | Execute if condition is True | `if x > 0:` |
| `elif` | Check another condition | `elif x < 0:` |
| `else` | Execute if all conditions False | `else:` |
| `for` | Iterate over sequence | `for i in range(5):` |
| `while` | Loop while condition True | `while x > 0:` |
| `break` | Exit loop immediately | `break` |
| `continue` | Skip to next iteration | `continue` |
| `pass` | Do nothing (placeholder) | `pass` |
| `match` | Pattern matching (3.10+) | `match value:` |
