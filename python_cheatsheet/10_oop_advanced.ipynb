## 1. Inheritance

```python
# Basic inheritance
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return "Some sound"
    
    def info(self):
        return f"I am {self.name}"

class Dog(Animal):  # Dog inherits from Animal
    def speak(self):  # Override parent method
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

# Create instances
dog = Dog("Buddy")
cat = Cat("Whiskers")

print(dog.info())    # Inherited method
print(dog.speak())   # Overridden method
print(cat.info())    # Inherited method
print(cat.speak())   # Overridden method
```

```python
# Using super() to call parent methods
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)  # Call parent constructor
        self.breed = breed
    
    def info(self):
        return f"{self.name} is a {self.age} year old {self.breed}"

dog = Dog("Buddy", 3, "Golden Retriever")
print(dog.info())
```

```python
# Multiple inheritance
class Flyable:
    def fly(self):
        return "Flying..."

class Swimmable:
    def swim(self):
        return "Swimming..."

class Duck(Flyable, Swimmable):
    def quack(self):
        return "Quack!"

duck = Duck()
print(duck.fly())    # From Flyable
print(duck.swim())   # From Swimmable
print(duck.quack())  # Own method

# Check inheritance
print(f"\nisinstance(duck, Flyable): {isinstance(duck, Flyable)}")
print(f"issubclass(Duck, Swimmable): {issubclass(Duck, Swimmable)}")
```

```python
# Method Resolution Order (MRO)
class A:
    def method(self):
        return "A"

class B(A):
    def method(self):
        return "B"

class C(A):
    def method(self):
        return "C"

class D(B, C):  # Diamond inheritance
    pass

d = D()
print(f"d.method(): {d.method()}")  # B (follows MRO)
print(f"MRO: {[cls.__name__ for cls in D.__mro__]}")
```

## 2. Polymorphism

```python
# Polymorphism through inheritance
class Shape:
    def area(self):
        raise NotImplementedError("Subclass must implement")

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        import math
        return math.pi * self.radius ** 2

# Polymorphic behavior
shapes = [Rectangle(4, 5), Circle(3), Rectangle(2, 8)]

for shape in shapes:
    print(f"{type(shape).__name__}: area = {shape.area():.2f}")
```

```python
# Duck typing (implicit polymorphism)
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Robot:
    def speak(self):
        return "Beep boop!"

# Function that works with any object having speak() method
def make_speak(entity):
    print(entity.speak())

# Duck typing - "If it walks like a duck and quacks like a duck..."
for entity in [Dog(), Cat(), Robot()]:
    make_speak(entity)
```

## 3. Encapsulation

```python
# Access modifiers (conventions)
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner           # Public
        self._account_type = "Savings"  # Protected (convention)
        self.__balance = balance     # Private (name mangling)
    
    # Public method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return True
        return False
    
    # Public method to access private attribute
    def get_balance(self):
        return self.__balance
    
    # Protected method (convention)
    def _calculate_interest(self):
        return self.__balance * 0.05

account = BankAccount("Alice", 1000)

# Public
print(f"Owner: {account.owner}")

# Protected (can access but shouldn't)
print(f"Account type: {account._account_type}")

# Private (name mangling)
# print(account.__balance)  # AttributeError!
print(f"Balance: {account.get_balance()}")  # Use getter

# Can still access via name mangling (not recommended)
print(f"Direct access: {account._BankAccount__balance}")
```

```python
# Using properties for encapsulation
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius
    
    @property
    def celsius(self):
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature below absolute zero!")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        return self._celsius * 9/5 + 32
    
    @fahrenheit.setter
    def fahrenheit(self, value):
        self.celsius = (value - 32) * 5/9

temp = Temperature(25)
print(f"Celsius: {temp.celsius}")
print(f"Fahrenheit: {temp.fahrenheit}")

temp.fahrenheit = 100
print(f"\nAfter setting to 100°F:")
print(f"Celsius: {temp.celsius:.2f}")
print(f"Fahrenheit: {temp.fahrenheit}")
```

## 4. Abstraction

```python
from abc import ABC, abstractmethod

# Abstract base class
class Shape(ABC):
    @abstractmethod
    def area(self):
        """Calculate area"""
        pass
    
    @abstractmethod
    def perimeter(self):
        """Calculate perimeter"""
        pass
    
    # Concrete method (can be inherited)
    def description(self):
        return f"I am a {type(self).__name__}"

# Cannot instantiate abstract class
# shape = Shape()  # TypeError!

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        import math
        return math.pi * self.radius ** 2
    
    def perimeter(self):
        import math
        return 2 * math.pi * self.radius

rect = Rectangle(4, 5)
circ = Circle(3)

for shape in [rect, circ]:
    print(shape.description())
    print(f"  Area: {shape.area():.2f}")
    print(f"  Perimeter: {shape.perimeter():.2f}")
```

```python
# Abstract properties
from abc import ABC, abstractmethod

class Database(ABC):
    @property
    @abstractmethod
    def connection_string(self):
        pass
    
    @abstractmethod
    def connect(self):
        pass
    
    @abstractmethod
    def disconnect(self):
        pass

class MySQLDatabase(Database):
    def __init__(self, host, port):
        self.host = host
        self.port = port
    
    @property
    def connection_string(self):
        return f"mysql://{self.host}:{self.port}"
    
    def connect(self):
        return f"Connected to {self.connection_string}"
    
    def disconnect(self):
        return "Disconnected from MySQL"

db = MySQLDatabase("localhost", 3306)
print(db.connection_string)
print(db.connect())
```

## 5. Composition vs Inheritance

```python
# Composition - "has a" relationship
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower
    
    def start(self):
        return "Engine started"
    
    def stop(self):
        return "Engine stopped"

class Wheels:
    def __init__(self, count=4):
        self.count = count
    
    def rotate(self):
        return f"{self.count} wheels rotating"

class Car:
    def __init__(self, model, horsepower):
        self.model = model
        self.engine = Engine(horsepower)  # Composition
        self.wheels = Wheels(4)           # Composition
    
    def start(self):
        return f"{self.model}: {self.engine.start()}"
    
    def drive(self):
        return f"{self.model}: {self.wheels.rotate()}"

car = Car("Tesla Model S", 670)
print(car.start())
print(car.drive())
print(f"Horsepower: {car.engine.horsepower}")
```

```python
# When to use Composition vs Inheritance
# 
# Inheritance ("is a"):
# - Dog IS AN Animal
# - Circle IS A Shape
# - Manager IS AN Employee
#
# Composition ("has a"):
# - Car HAS AN Engine
# - House HAS Rooms
# - Computer HAS a CPU

# Example showing when composition is preferred
class Logger:
    def log(self, message):
        print(f"[LOG] {message}")

class FileHandler:
    def save(self, data):
        print(f"Saving: {data}")

# Using composition instead of multiple inheritance
class Application:
    def __init__(self):
        self.logger = Logger()
        self.file_handler = FileHandler()
    
    def process(self, data):
        self.logger.log(f"Processing {data}")
        self.file_handler.save(data)
        self.logger.log("Done processing")

app = Application()
app.process("user data")
```

## 6. Mixins

```python
# Mixins - add functionality without full inheritance
class JSONMixin:
    def to_json(self):
        import json
        return json.dumps(self.__dict__)

class LoggableMixin:
    def log(self, message):
        print(f"[{type(self).__name__}] {message}")

class ComparableMixin:
    def __eq__(self, other):
        return self.__dict__ == other.__dict__

class Person(JSONMixin, LoggableMixin, ComparableMixin):
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Alice", 25)
p2 = Person("Alice", 25)

print(p1.to_json())      # From JSONMixin
p1.log("Hello!")         # From LoggableMixin
print(f"p1 == p2: {p1 == p2}")  # From ComparableMixin
```

## OOP Summary

| Concept | Description | Key Points |
|---------|-------------|------------|
| Inheritance | Child class inherits from parent | Use `super()` to call parent |
| Polymorphism | Same interface, different behavior | Duck typing in Python |
| Encapsulation | Hide internal details | `_protected`, `__private` |
| Abstraction | Define interface without implementation | Use `ABC` and `@abstractmethod` |
